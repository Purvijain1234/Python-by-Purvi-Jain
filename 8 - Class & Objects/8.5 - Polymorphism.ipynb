## 5. Polymorphism

**Polymorphism** means "many forms" - the same interface (method name) works differently depending on the object it's called on. It lets you write generic code that works with any object that has the right method.

### Two main forms in Python
| Form | How it works |
|------|--------------|
| **Method overriding** | Subclass redefines a parent method |
| **Duck typing** | If it has the right method, it works - no need to share a parent class |

### Duck typing
> *"If it walks like a duck and quacks like a duck, it's a duck."*

Python doesn't check the type of an object - it just calls the method and sees what happens. This makes Python extremely flexible and allows **protocols** (informal interfaces) like iteration, context managers, and comparison operators.

## Method overriding polymorphism

```python
# ---- Method overriding polymorphism ----
class Shape:
    def area(self):        raise NotImplementedError
    def perimeter(self):   raise NotImplementedError
    def describe(self):
        return (f"{self.__class__.__name__}: "
                f"area={self.area():.2f}, perimeter={self.perimeter():.2f}")

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w, self.h = w, h
    def area(self):        return self.w * self.h
    def perimeter(self):   return 2 * (self.w + self.h)

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):        return math.pi * self.r ** 2
    def perimeter(self):   return 2 * math.pi * self.r

class Triangle(Shape):
    def __init__(self, a, b, c):
        self.a, self.b, self.c = a, b, c
    def perimeter(self):
        return self.a + self.b + self.c
    def area(self):
        s = self.perimeter() / 2
        return math.sqrt(s * (s-self.a) * (s-self.b) * (s-self.c))

import math

shapes = [Rectangle(4, 6), Circle(5), Triangle(3, 4, 5)]

# Same method call — different behaviour per object
print("----- Polymorphic describe() -----")
for shape in shapes:
    print(f"  {shape.describe()}")

# Generic functions that work with any Shape
def total_area(shapes):
    return sum(s.area() for s in shapes)

def largest_shape(shapes):
    return max(shapes, key=lambda s: s.area())

print(f"\nTotal area:   {total_area(shapes):.2f}")
```

## Duck typing

```python
# ---- Duck typing ----
# These classes share NO parent — they just all have a .draw() method
class PDFRenderer:
    def draw(self, text):
        return f"[PDF] rendering: {text!r}"

class HTMLRenderer:
    def draw(self, text):
        return f"<p>{text}</p>"

class ConsoleRenderer:
    def draw(self, text):
        return f">> {text}"

def render_document(renderer, paragraphs):
    """Works with any object that has a .draw() method."""
    for p in paragraphs:
        print(renderer.draw(p))

content = ["Title: Python OOP", "Introduction paragraph.", "Conclusion."]

for renderer in [PDFRenderer(), HTMLRenderer(), ConsoleRenderer()]:
    print(f"--- {renderer.__class__.__name__} ---")
    render_document(renderer, content)
    print()
```
