## 1. Classes & `__init__`

A **class** defines the structure and behaviour that all objects of that type will share.

```python
class ClassName:               # PascalCase naming convention
    """Class docstring."""

    def __init__(self, param1, param2):
        """Constructor - called automatically when an object is created."""
        self.param1 = param1   # instance attribute
        self.param2 = param2
```

### Key concepts
| Term | Meaning |
|------|---------|
| `class` | Keyword to define a class |
| `__init__` | Constructor — initialises a new instance |
| `self` | Reference to the **current instance** - always the first parameter |
| Instance | A specific object created from the class (`obj = ClassName()`) |
| Attribute | Data stored on an object (`self.name`) |
| Method | Function defined inside a class |

> 💡 `self` is just a convention - Python passes the instance automatically. You could name it anything, but **always use `self`**.

````python
# Simplest possible class
class Dog:
    """Represents a dog."""
    pass

d = Dog()                     # create an instance
print(type(d))                # <class '__main__.Dog'>
print(isinstance(d, Dog))     # True
````

````python
# Class with __init__ and attributes
class Student:
    """Represents a student with a name, age, and grade."""

    def __init__(self, name, age, grade):
        self.name  = name       # instance attributes set via self
        self.age   = age
        self.grade = grade

    def introduce(self):
        """Print a self-introduction."""
        print(f"Hi, I'm {self.name}, {self.age} years old, grade {self.grade}.")

# Creating multiple independent instances
s1 = Student("Purvi",   21, "A")
s2 = Student("Alice",   20, "B")
s3 = Student("Charlie", 22, "A")

s1.introduce()
s2.introduce()
s3.introduce()
````

````python
# Accessing and modifying attributes directly
print(f"Before: {s1.name}, grade {s1.grade}")
s1.grade = "A+"               # direct attribute update
s1.email = "purvi@ex.com"     # add a new attribute dynamically
print(f"After : {s1.name}, grade {s1.grade}, email {s1.email}")
````

````python
# Each instance is independent
print(f"s1 is s2: {s1 is s2}")   # False — different objects
print(f"s1 id: {id(s1)}")
print(f"s2 id: {id(s2)}")
````

````python
# __init__ with default values
class Counter:
    """A simple counter that starts at a given value."""

    def __init__(self, start=0, step=1):
        self.value = start
        self.step  = step

    def increment(self):
        self.value += self.step

    def reset(self):
        self.value = 0

c1 = Counter()          # start=0, step=1
c2 = Counter(10, 5)     # start=10, step=5

c1.increment(); c1.increment(); c1.increment()
c2.increment(); c2.increment()

print(f"c1.value = {c1.value}")   # 3
print(f"c2.value = {c2.value}")   # 20
````
