## 4. Inheritance

**Inheritance** lets a new class (**child / subclass**) reuse and extend the behaviour of an existing class (**parent / superclass**). This models **"is-a"** relationships.

```python
class Child(Parent):
    def __init__(self, ...):
        super().__init__(...)   # call parent constructor
        # child-specific setup
```

### Key rules
| Concept | Detail |
|---------|--------|
| `super()` | Proxy to the parent class — call parent methods without hardcoding the class name |
| Method override | Child defines a method with the same name as parent → child version is used |
| `isinstance(obj, cls)` | True if `obj` is an instance of `cls` **or any subclass** |
| `issubclass(A, B)` | True if class A is a subclass of B |
| Multiple inheritance | `class C(A, B):` — Python uses **MRO (C3 linearisation)** to resolve order |

### Single vs Multiple inheritance
```
Single:   Animal → Dog
Multi:    Flyable, Swimmable → Duck
Multi-level: Animal → Mammal → Dog
```

````python
# Base class
class Animal:
    """Base class for all animals."""

    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def speak(self):
        """Override in subclass."""
        raise NotImplementedError(f"{self.__class__.__name__} must implement speak()")

    def info(self):
        return f"{self.__class__.__name__} | name={self.name!r}, age={self.age}"

    def __repr__(self):
        return self.info()
````

## Single Inheritance

````python
# Single inheritance
class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)    # call Animal.__init__
        self.breed = breed

    def speak(self):
        return f"{self.name} says: Woof!"

    def fetch(self, item):
        return f"{self.name} fetched the {item}!"

    def info(self):
        return super().info() + f", breed={self.breed!r}"


class Cat(Animal):
    def __init__(self, name, age, indoor=True):
        super().__init__(name, age)
        self.indoor = indoor

    def speak(self):
        return f"{self.name} says: Meow!"

    def info(self):
        return super().info() + f", indoor={self.indoor}"


class GuideDog(Dog):
    """Multi-level: GuideDog → Dog → Animal."""

    def __init__(self, name, age, breed, owner):
        super().__init__(name, age, breed)
        self.owner = owner

    def guide(self):
        return f"{self.name} is guiding {self.owner} safely."


dog   = Dog("Bruno", 3, "Labrador")
cat   = Cat("Whiskers", 5)
guide = GuideDog("Rex", 4, "German Shepherd", "Purvi")

print(dog.info())
print(cat.info())
print(guide.info())
````

````python
print(dog.speak())
print(cat.speak())
print(dog.fetch("ball"))
print(guide.guide())
````

````python
# isinstance / issubclass
print(f"dog  isinstance Animal: {isinstance(dog,   Animal)}")
print(f"guide isinstance Dog:   {isinstance(guide, Dog)}")
print(f"guide isinstance Animal:{isinstance(guide, Animal)}")
print(f"GuideDog subclass Dog:  {issubclass(GuideDog, Dog)}")
````

## Multiple Inheritance

````python
# Multiple inheritance + MRO
class Flyable:
    def move(self): return "flying"

class Swimmable:
    def move(self): return "swimming"

class Duck(Flyable, Swimmable, Animal):
    def __init__(self, name, age):
        Animal.__init__(self, name, age)
    def speak(self):
        return f"{self.name} says: Quack!"

duck = Duck("Donald", 2)
print(duck.speak())
print(f"duck.move() → {duck.move()}")   # Flyable comes first in MRO
print(f"MRO: {[c.__name__ for c in Duck.__mro__]}")
````
