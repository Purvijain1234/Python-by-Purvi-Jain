## 2. Instance vs Class Variables

Python has two kinds of variables inside a class:

| | Instance Variable | Class Variable |
|-|-------------------|----------------|
| Defined in | `__init__` via `self.x` | Class body (outside any method) |
| Belongs to | Each **individual object** | The **class itself** (shared by all instances) |
| Access | `obj.x` | `ClassName.x` or `obj.x` |
| Unique per instance | ✅ Yes | ❌ No — shared |

### The shadowing trap
If you assign to `self.x` when a class variable `x` already exists, Python creates a **new instance variable** that **shadows** the class variable for that object - the class variable is not modified.

### Typical uses for class variables
- Counting how many instances exist.
- Storing constants or defaults shared by all instances.
- Registry / lookup tables shared across the class.

```python
class BankAccount:
    """Bank account with instance and class variables."""

    bank_name    = "Python Bank"   # class variable — shared by all
    interest_rate = 0.05           # class variable
    _total_accounts = 0            # class variable — instance counter

    def __init__(self, owner, balance=0):
        self.owner   = owner       # instance variable
        self.balance = balance     # instance variable
        BankAccount._total_accounts += 1   # update class variable via class name

    def annual_interest(self):
        # mixes class variable and instance variable
        return self.balance * BankAccount.interest_rate

    @classmethod
    def total_accounts(cls):
        """Return total number of accounts created."""
        return cls._total_accounts

# Class variable access before any instance
print(f"Bank: {BankAccount.bank_name}")
print(f"Rate: {BankAccount.interest_rate * 100}%")
print(f"Accounts so far: {BankAccount.total_accounts()}")
```

```python
# Creating instances
acc1 = BankAccount("Purvi",  50000)
acc2 = BankAccount("Alice",  30000)
acc3 = BankAccount("Bob",    10000)

print(f"Accounts after creating 3: {BankAccount.total_accounts()}")
print(f"acc1 interest: ₹{acc1.annual_interest():,.0f}")
print(f"acc2 interest: ₹{acc2.annual_interest():,.0f}")
```

```python
# Changing class variable affects ALL instances
print("Changing interest rate to 7%...")
BankAccount.interest_rate = 0.07
print(f"acc1 new interest: ₹{acc1.annual_interest():,.0f}")
print(f"acc2 new interest: ₹{acc2.annual_interest():,.0f}")
```

```python
# Shadowing trap - assigning to instance creates a NEW instance variable
print("Shadowing demo:")
acc1.interest_rate = 0.10   # creates instance variable, does NOT change class var
print(f"acc1.interest_rate (instance): {acc1.interest_rate}")   # 0.10
print(f"acc2.interest_rate (class):    {acc2.interest_rate}")   # 0.07
print(f"BankAccount.interest_rate:     {BankAccount.interest_rate}")  # 0.07
```

```python
# Check instance __dict__ vs class __dict__
print("acc1.__dict__:", acc1.__dict__)        # includes shadowed interest_rate
print("acc2.__dict__:", acc2.__dict__)        # does NOT have interest_rate
```
