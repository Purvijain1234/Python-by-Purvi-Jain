## 3. Methods & `self`

A **method** is a function defined inside a class. Python has three kinds:

| Method type | Decorator | First param | Accesses |
|-------------|-----------|-------------|----------|
| **Instance method** | _(none)_ | `self` | Instance attributes & class attributes |
| **Class method** | `@classmethod` | `cls` | Class attributes only (not instance) |
| **Static method** | `@staticmethod` | _(none)_ | Neither - utility function in the class namespace |

### When to use which
- **Instance method** - the default; needs `self` data to do its work.
- **Class method** - factory constructors, alternative constructors, counters.
- **Static method** - helper/utility logic that belongs conceptually in the class but doesn't need `self` or `cls`.

```python
import math
from datetime import date

class Circle:
    """Demonstrates all three method types."""

    PI = math.pi                  # class variable

    def __init__(self, radius):
        self.radius = radius      # instance variable

    # --- Instance methods (most common) ---
    def area(self):
        """Return the area of the circle."""
        return Circle.PI * self.radius ** 2

    def circumference(self):
        """Return the circumference."""
        return 2 * Circle.PI * self.radius

    def scale(self, factor):
        """Scale the radius by factor and return self (fluent interface)."""
        self.radius *= factor
        return self            # allows method chaining

    # --- Class method ---
    @classmethod
    def from_diameter(cls, diameter):
        """Alternative constructor — create Circle from diameter."""
        return cls(diameter / 2)

    @classmethod
    def unit_circle(cls):
        """Return a circle with radius 1."""
        return cls(1)

    # --- Static method ---
    @staticmethod
    def is_valid_radius(r):
        """Return True if r is a positive number."""
        return isinstance(r, (int, float)) and r > 0

    def __repr__(self):
        return f"Circle(radius={self.radius:.2f})"
```

```python
# Instance methods
c = Circle(5)
print(f"Circle:        {c}")
print(f"Area:          {c.area():.4f}")
print(f"Circumference: {c.circumference():.4f}")
```

```python
# Method chaining (scale returns self)
c2 = Circle(3).scale(2).scale(1.5)
print(f"After chained scaling: {c2}")
```

```python
# Class methods - alternative constructors
c3 = Circle.from_diameter(10)
c4 = Circle.unit_circle()
print(f"from_diameter(10): {c3}")
print(f"unit_circle():     {c4}")
```

```python
# Static method - utility, no instance needed
for val in [5, -3, 0, "abc", 2.5]:
    print(f"  is_valid_radius({val!r}): {Circle.is_valid_radius(val)}")
```

```python
# Real-world class method: parse from string
class Person:
    def __init__(self, name, birth_year):
        self.name       = name
        self.birth_year = birth_year

    @classmethod
    def from_string(cls, data_str):
        """Parse 'Name-YYYY' format."""
        name, year = data_str.split("-")
        return cls(name.strip(), int(year.strip()))

    @staticmethod
    def is_adult(age):
        return age >= 18

    @property
    def age(self):
        return date.today().year - self.birth_year

    def __repr__(self):
        return f"Person(name={self.name!r}, age={self.age})"

p1 = Person("Purvi", 2003)
p2 = Person.from_string("Alice - 1995")
print(p1)
print(p2)
print(f"Is p1 adult: {Person.is_adult(p1.age)}")
```
