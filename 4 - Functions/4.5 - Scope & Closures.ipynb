## 5. Scope & Closures

**Scope** defines where a variable is visible. Python uses the **LEGB rule** to look up names:

```
L - Local      → inside the current function
E - Enclosing  → in any enclosing function (for closures)
G - Global     → at the module (file) level
B - Built-in   → Python's built-in names (len, print, range…)
```

Python searches from **L → E → G → B** and uses the first match found.

### `global` and `nonlocal`
| Keyword | Used inside | Allows writing to |
|---------|-------------|-------------------|
| `global x` | any function | module-level `x` |
| `nonlocal x` | nested function | enclosing function's `x` |

### Closures
A **closure** is a nested function that **remembers variables from its enclosing scope** even after that scope has finished executing. They are the foundation of decorators and factory functions.

## Scope

````python
x = "global"

def show_scope():
    x = "local"       # creates a NEW local variable — does not touch global x
    print(f"Inside function: x = {x!r}")

show_scope()
print(f"Outside function: x = {x!r}")  # global x unchanged
````

````python
# global keyword — modify a global variable from inside a function
counter = 0

def increment():
    global counter          # declare intent to modify global
    counter += 1

increment()
increment()
increment()
print(f"counter = {counter}")   # 3
````

## Clousers

````python
# Factory function - returns a customised function
def make_multiplier(factor):
    """Return a function that multiplies its argument by factor."""
    def multiply(n):             # inner function
        return n * factor        # 'factor' is from the enclosing scope
    return multiply              # return the inner function itself

double = make_multiplier(2)
triple = make_multiplier(3)
times10 = make_multiplier(10)

print(double(5))    # 10
print(triple(5))    # 15
print(times10(7))   # 70

# Each closure remembers its own 'factor'
print(double.__closure__[0].cell_contents)   # 2
print(triple.__closure__[0].cell_contents)   # 3
````

````python
# nonlocal - modify enclosing function's variable
def make_counter(start=0):
    count = start
    def increment(step=1):
        nonlocal count       # refer to enclosing 'count'
        count += step
        return count
    return increment

cnt = make_counter(10)
print(cnt())     # 11
print(cnt())     # 12
print(cnt(5))    # 17
````
