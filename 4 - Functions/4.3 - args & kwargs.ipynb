## 3. `*args` & `**kwargs`

Sometimes you don't know in advance how many arguments a function will receive. Python provides two special syntaxes for this:

| Syntax | Name | Collects into | Example call |
|--------|------|---------------|--------------|
| `*args` | Arbitrary positional | **tuple** | `f(1, 2, 3)` |
| `**kwargs` | Arbitrary keyword | **dict** | `f(a=1, b=2)` |

### Parameter order rule (must follow this exactly)
```python
def f(positional, /, normal, *args, keyword_only, **kwargs):
#     pos-only       regular   var-pos  kw-only      var-kw
```

### Unpacking at the call site
The `*` and `**` operators also work in reverse - you can **unpack** a list/tuple or dict into a function call:
```python
args = [1, 2, 3]
f(*args)           # same as f(1, 2, 3)

kwargs = {"a": 1, "b": 2}
f(**kwargs)        # same as f(a=1, b=2)
```

````python
# *args - variable number of positional arguments
def total(*args):
    """Sum any number of numbers."""
    print(f"  args = {args}   type = {type(args)}")
    return sum(args)

print(total(1, 2))
print(total(10, 20, 30, 40))
print(total())              # empty tuple - returns 0
````

````python
# **kwargs - variable number of keyword arguments
def display_info(**kwargs):
    """Print any key-value info."""
    print(f"  kwargs = {kwargs}")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

display_info(name="Purvi", age=21, city="Delhi")
````

````python
# Combining all parameter types
def full_example(required, default="hi", *args, kw_only, **kwargs):
    print(f"required  = {required}")
    print(f"default   = {default}")
    print(f"args      = {args}")
    print(f"kw_only   = {kw_only}")
    print(f"kwargs    = {kwargs}")

full_example("hello", "world", 1, 2, 3, kw_only="must pass", extra="yes")
````

````python

# Real-world use: flexible logger
def log(level, *messages, sep=" | ", **meta):
    joined = sep.join(str(m) for m in messages)
    tags   = " ".join(f"[{k}={v}]" for k, v in meta.items())
    print(f"[{level.upper()}] {joined} {tags}")

log("info",  "Server started", "Port 8000")
log("error", "DB connection failed", sep=" >> ", user="admin", code=500)
````

````python
# Unpacking at the call site (*list, **dict)
def greet(first, last, title="Mr"):
    print(f"Hello, {title} {first} {last}!")

names  = ["John", "Doe"]
extras = {"title": "Dr"}

greet(*names)              # unpack list as positional args
greet(*names, **extras)    # unpack both
````
