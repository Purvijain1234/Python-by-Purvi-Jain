## 1. Defining Functions

Use the `def` keyword to define a function. The function body is indented by 4 spaces.

```python
def function_name(parameters):
    """Docstring - what this function does."""
    # body
    return result
```

### Anatomy of a function
| Part | Role |
|------|------|
| `def` | Keyword that declares a function |
| `function_name` | snake_case name describing what it does |
| `parameters` | Input variables (optional) |
| `"""docstring"""` | Human-readable description (optional but recommended) |
| `return` | Sends a value back to the caller (optional — returns `None` if omitted) |

> 💡 A function is only **defined** when `def` runs. It is **executed** only when you **call** it with `()`.

````python
# Simplest function - no parameters, no return value
def greet():
    """Print a greeting message."""
    print("Hello, World!")

greet()          # call the function
greet()          # call it again - reusable!

print()
````

````python
# Function with a parameter
def greet_user(name):
    """Print a personalised greeting."""
    print(f"Hello, {name}! Welcome to Python.")

greet_user("Purvi")
greet_user("Alice")
````

````python
# Function with a return value
def square(n):
    """Return the square of n."""
    return n ** 2

result = square(5)
print(f"square(5) = {result}")
print(f"square(9) = {square(9)}")   # use return value directly
````

````python
# Functions are first-class objects - assignable, passable
my_func = square          # assign function to a variable (no parentheses!)
print(my_func(4))         # 16
````

````python
# Accessing the docstring
print(square.__doc__)
help(square)              # formatted version
````
