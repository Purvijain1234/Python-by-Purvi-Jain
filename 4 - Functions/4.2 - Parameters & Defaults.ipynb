## 2. Parameters & Defaults

Python supports four kinds of function parameters:

| Kind | Syntax | Notes |
|------|--------|-------|
| Positional | `def f(a, b)` | Must be passed in order |
| Default | `def f(a, b=10)` | Optional - uses default if not supplied |
| Keyword (at call) | `f(b=5, a=3)` | Order doesn't matter when using keyword |
| Positional-only (3.8+) | `def f(a, b, /, c)` | `a`, `b` cannot be passed as keyword |

### Rules for default parameters
1. Parameters **with** defaults must come **after** parameters **without** defaults.
2. ⚠️ **Never use a mutable object (list, dict) as a default value** - it is shared across all calls. Use `None` and create inside the function body instead.

### Argument vs Parameter
- **Parameter** - the variable name in the function definition (`name` in `def greet(name)`)
- **Argument** - the actual value passed when calling (`"Purvi"` in `greet("Purvi")`)

```python
# Positional parameters
def add(a, b):
    return a + b

print(add(3, 7))         # positional — order matters
print(add(b=7, a=3))     # keyword — order doesn't matter
```

```python
# Default parameter values
def power(base, exponent=2):    # exponent defaults to 2
    """Return base raised to exponent."""
    return base ** exponent

print(power(3))        # 9   - uses default exponent=2
print(power(3, 3))     # 27  - overrides default
print(power(2, 10))    # 1024
```

```python
# Mixing positional and keyword arguments
def describe_pet(name, animal="dog", age=1):
    print(f"{name} is a {age}-year-old {animal}.")

describe_pet("Bruno")                      # uses all defaults
describe_pet("Whiskers", "cat")            # positional override
describe_pet("Nemo", age=3, animal="fish") # keyword override, any order
```

```python
# ⚠️ Mutable default argument - WRONG pattern
def bad_append(item, lst=[]):   # list is shared across calls!
    lst.append(item)
    return lst

print(bad_append(1))   # [1]     - looks fine
print(bad_append(2))   # [1, 2]  - BUG: same list reused!
print(bad_append(3))   # [1, 2, 3]
```

```python
# ✅ Correct pattern - use None as sentinel
def good_append(item, lst=None):
    if lst is None:
        lst = []        # fresh list every call
    lst.append(item)
    return lst

print(good_append(1))   # [1]
print(good_append(2))   # [2]  - correctly independent
print(good_append(3))   # [3]
```
