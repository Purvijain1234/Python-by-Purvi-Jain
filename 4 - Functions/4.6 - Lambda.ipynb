## 6. Lambda Functions

A **lambda** is a small, **anonymous** (nameless) function written in a single expression. It is syntactic sugar for a simple `def` function.

```python
lambda parameters: expression
```

### Lambda vs def
| | `lambda` | `def` |
|-|----------|-------|
| Name | Anonymous (usually) | Named |
| Body | Single expression | Any code |
| `return` | Implicit | Explicit |
| Docstring | ❌ | ✅ |
| Debugging | Harder | Easier |

### When to use lambdas
- As the `key=` argument for `sorted()`, `min()`, `max()`.
- As a short callback passed to `map()`, `filter()`, or GUI event handlers.
- Anywhere the function is used **exactly once** and has a trivial body.

> 💡 If you find yourself assigning a lambda to a variable (`f = lambda x: x*2`), just use `def` instead - it's cleaner and more debuggable.

````python
# Basic lambda
square = lambda x: x ** 2
print(square(5))     # 25

add = lambda a, b: a + b
print(add(3, 7))     # 10

# Equivalent def (prefer def when naming the function)
def square_def(x):
    return x ** 2
````

````python
# ✅ Best use: key= argument in sorted / min / max
students = [
    {"name": "Charlie", "score": 72},
    {"name": "Alice",   "score": 95},
    {"name": "Bob",     "score": 88},
]

by_score = sorted(students, key=lambda s: s["score"])
print("By score (asc):", [s["name"] for s in by_score])

by_score_desc = sorted(students, key=lambda s: s["score"], reverse=True)
print("By score (desc):", [s["name"] for s in by_score_desc])

top = max(students, key=lambda s: s["score"])
print(f"Top student: {top['name']} ({top['score']})")

print()

# Sorting by multiple criteria (tuple as key)
people = [("Alice", 30), ("Bob", 25), ("Charlie", 30), ("Dave", 25)]
# Sort by age, then alphabetically by name
sorted_people = sorted(people, key=lambda p: (p[1], p[0]))
print("Multi-key sort:", sorted_people)
````

````python
# Lambda inside a function (conditional expression)
operations = {
    "add":      lambda a, b: a + b,
    "subtract": lambda a, b: a - b,
    "multiply": lambda a, b: a * b,
    "divide":   lambda a, b: a / b if b != 0 else "undefined",
}

for op_name, op_func in operations.items():
    print(f"  10 {op_name} 3 = {op_func(10, 3)}")
````
