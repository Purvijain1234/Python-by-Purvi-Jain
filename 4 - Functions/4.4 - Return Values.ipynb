## 4. Return Values

The `return` statement does two things:
1. **Sends a value** back to the caller.
2. **Immediately exits** the function - any code after `return` in the same branch is unreachable.

### Key rules
| Situation | What gets returned |
|-----------|--------------------|
| `return value` | that value |
| `return` (bare) | `None` |
| No `return` at all | `None` |
| `return a, b, c` | tuple `(a, b, c)` |

### Early return (guard clause)
Returning early when invalid input is detected keeps the main logic clean and flat - the same guard-clause idea from the Control Flow notebook, now inside functions.

```python
# Basic return
def celsius_to_fahrenheit(c):
    """Convert Celsius to Fahrenheit."""
    return c * 9 / 5 + 32

print(celsius_to_fahrenheit(0))    # 32.0
print(celsius_to_fahrenheit(100))  # 212.0
print(celsius_to_fahrenheit(37))   # 98.6
```

```python
# No return → returns None
def do_nothing():
    pass

result = do_nothing()
print(f"do_nothing() returned: {result!r}")   # None
```

```python
# Multiple return values (actually a tuple)
def stats(numbers):
    """Return (minimum, maximum, average) of a list."""
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

data = [4, 7, 2, 9, 1, 5, 8]
lo, hi, avg = stats(data)          # unpack the tuple
print(f"min={lo}, max={hi}, avg={avg:.2f}")

raw = stats(data)                  # keep as tuple
print(f"raw return: {raw}")
```

```python
# Early return - guard clause pattern
def divide(a, b):
    """Return a / b, or None if b is zero."""
    if b == 0:
        print("  Cannot divide by zero.")
        return None        # exit early
    return a / b           # happy path

print(divide(10, 2))   # 5.0
print(divide(10, 0))   # None (after warning)
```

```python
# Returning different types based on logic
def classify_bmi(weight_kg, height_m):
    """Classify BMI into a category string."""
    if height_m <= 0:
        return "invalid input"
    bmi = weight_kg / height_m ** 2
    if bmi < 18.5:
        return "underweight"
    elif bmi < 25:
        return "normal"
    elif bmi < 30:
        return "overweight"
    else:
        return "obese"

for w, h in [(50, 1.70), (70, 1.70), (90, 1.70), (120, 1.70)]:
    print(f"  {w}kg / {h}m → {classify_bmi(w, h)}")
```
