## 1. `try` / `except` / `else`

The full syntax of a try block:

```python
try:
    # code that MIGHT raise an exception
except SomeError:
    # runs ONLY if SomeError was raised in try
except (TypeError, ValueError) as e:
    # catch multiple exception types at once
    # 'e' holds the exception object
else:
    # runs ONLY if NO exception was raised in try
```

### How Python processes it
1. Execute the `try` block line by line.
2. If an exception is raised → jump to the **first matching** `except`.
3. If no exception → run `else` (skipped if an exception occurred).
4. After handling, continue with the code after the entire try-block.

### Rules
| Rule | Detail |
|------|--------|
| Order of `except` clauses | More specific exceptions **first**, broader ones last |
| Bare `except:` | Catches **everything** including `SystemExit` - avoid it |
| `except Exception:` | Catches all normal exceptions - acceptable as a catch-all |
| `as e` | Binds the exception instance so you can inspect its message |
| `else` | Runs only on success - keeps "happy path" separate from error handling |

````python
# Basic try / except
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")

print()

# Catching the exception object with 'as e'
try:
    number = int("hello")        # ValueError
except ValueError as e:
    print(f"ValueError caught: {e}")
    print(f"Exception type:   {type(e).__name__}")
````

````python
# Multiple except clauses
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print(f"  [{a}/{b}] Error: division by zero")
        return None
    except TypeError as e:
        print(f"  [{a}/{b}] Error: wrong type — {e}")
        return None
    else:
        # Only runs when no exception occurred
        print(f"  [{a}/{b}] Success: {result:.4f}")
        return result

safe_divide(10, 4)
safe_divide(10, 0)
safe_divide(10, "x")
````

````python
# Catching multiple exceptions in one clause
def parse_input(value):
    try:
        return int(value)
    except (ValueError, TypeError) as e:
        print(f"  parse_input({value!r}): {type(e).__name__}: {e}")
        return 0

print(parse_input("42"))    # 42
print(parse_input("abc"))   # ValueError
print(parse_input(None))    # TypeError
````

````python
# else clause — separates happy path from error handling
filenames = ["data.csv", "missing.csv", "notes.txt"]
for fname in filenames:
    try:
        f = open(fname, "r")
    except FileNotFoundError:
        print(f"  {fname}: not found")
    else:
        content = f.read()
        f.close()
        print(f"  {fname}: read {len(content)} bytes")
````
