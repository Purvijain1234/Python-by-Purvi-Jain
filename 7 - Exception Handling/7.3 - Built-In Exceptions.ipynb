## 3. Common Built-in Exceptions

Python has over 60 built-in exceptions. Here are the ones you'll encounter daily:

| Exception | When it occurs | Example |
|-----------|---------------|--------|
| `ValueError` | Right type, wrong value | `int("abc")` |
| `TypeError` | Wrong type for operation | `"2" + 2` |
| `NameError` | Variable not defined | `print(undefined_var)` |
| `IndexError` | List index out of range | `lst[99]` on a 3-item list |
| `KeyError` | Dict key doesn't exist | `d["missing"]` |
| `AttributeError` | Object has no such attribute | `None.upper()` |
| `ZeroDivisionError` | Division or modulo by zero | `10 / 0` |
| `FileNotFoundError` | File/directory doesn't exist | `open("ghost.txt")` |
| `PermissionError` | No OS permission | Write to read-only file |
| `ImportError` | Module not found | `import nonexistent` |
| `StopIteration` | Iterator exhausted | `next()` on empty iterator |
| `RecursionError` | Max recursion depth exceeded | Infinite recursive call |
| `MemoryError` | Out of memory | Creating huge object |
| `OSError` | OS-level failures | Disk full, bad file descriptor |
| `RuntimeError` | Generic runtime issue | Catch-all for various errors |

```python
# Helper to demonstrate each exception cleanly
def demo(label, code_str):
    try:
        exec(code_str)
    except Exception as e:
        print(f"  {label:25s} → {type(e).__name__}: {e}")

print("=== Common Exceptions ===")
demo("int('abc')",             "int('abc')")
demo("'2' + 2",                "'2' + 2")
demo("undefined_var",          "print(undefined_var)")
demo("[][99]",                 "[][99]")
demo("{} ['key']",             "{}['key']")
demo("None.upper()",           "None.upper()")
demo("10 / 0",                 "10 / 0")
demo("open('ghost.txt')",      "open('ghost.txt')")
demo("import nonexistent",     "import nonexistent_xyz")
demo("next(iter([]))",         "next(iter([]))")
```

```python
# --- Inspecting exception details ---
try:
    data = {"user": {"name": "Alice"}}
    age = data["user"]["age"]          # KeyError
except KeyError as e:
    print(f"KeyError — missing key: {e}")
    print(f"  args:  {e.args}")
```

```python
# --- IndexError with helpful message ---
def safe_get(lst, index, default=None):
    """Return lst[index] or default if out of range."""
    try:
        return lst[index]
    except IndexError:
        return default

items = [10, 20, 30]
print(safe_get(items, 1))    # 20
print(safe_get(items, 9))    # None
print(safe_get(items, -1))   # 30 (valid negative index)
```

```python
# --- AttributeError guard ---
def get_upper(value):
    try:
        return value.upper()
    except AttributeError:
        return str(value).upper()

print(get_upper("hello"))   # HELLO
print(get_upper(42))        # 42 → '42' → '42' (no-op on digits)
print(get_upper(None))      # NONE
```

```python
# --- RecursionError ---
def infinite_recurse(n):
    return infinite_recurse(n + 1)

try:
    infinite_recurse(0)
except RecursionError as e:
    print(f"RecursionError caught: {str(e)[:50]}")
```
