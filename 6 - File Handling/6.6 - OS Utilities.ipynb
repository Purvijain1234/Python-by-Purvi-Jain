## 6. `os.path` Utilities

`os.path` is a sub-module of `os` that provides functions for working with **file and directory paths** in a cross-platform way (Windows uses `\`, Unix uses `/` — `os.path` handles both).

### Most-used functions
| Function | Returns | Example |
|----------|---------|--------|
| `os.path.join(a, b, ...)` | Safe joined path | `'home/user/file.txt'` |
| `os.path.exists(p)` | `bool` — path exists? | `True` / `False` |
| `os.path.isfile(p)` | `bool` — is a file? | `True` |
| `os.path.isdir(p)` | `bool` — is a directory? | `False` |
| `os.path.basename(p)` | Filename only | `'file.txt'` |
| `os.path.dirname(p)` | Directory only | `'/home/user'` |
| `os.path.splitext(p)` | `(name, ext)` tuple | `('file', '.txt')` |
| `os.path.abspath(p)` | Absolute path | `'/home/user/file.txt'` |
| `os.path.getsize(p)` | File size in bytes | `1024` |

> 💡 Use `os.path.join()` instead of string concatenation — it ensures the correct separator for the OS.

```python
import os

# Building paths safely (works on any OS)
path = os.path.join('project', 'data', 'raw', 'input.csv')
print('Joined path :', path)

# Absolute path of the current directory
print('Absolute CWD:', os.path.abspath('.'))

# Expand ~ (home directory shortcut)
home_file = os.path.expanduser('~/Documents/notes.txt')
print('Expanded    :', home_file)
```

```python
import os

# Inspecting a path — decompose it
p = '/home/user/projects/myapp/data/report_2024.csv'

print('basename  :', os.path.basename(p))    # report_2024.csv
print('dirname   :', os.path.dirname(p))     # /home/user/projects/myapp/data
print('splitext  :', os.path.splitext(p))    # ('/home/.../report_2024', '.csv')

name, ext = os.path.splitext(os.path.basename(p))
print('name only :', name)                   # report_2024
print('extension :', ext)                    # .csv

# split() — splits into (head, tail)
print('split     :', os.path.split(p))       # ('/home/.../data', 'report_2024.csv')
```

```python
import os

# Existence checks — always do these before reading/writing!
print('demo.txt exists?  ', os.path.exists('demo.txt'))     # True
print('missing.txt exists?', os.path.exists('missing.txt')) # False
print('demo.txt is file? ', os.path.isfile('demo.txt'))     # True
print('demo.txt is dir?  ', os.path.isdir('demo.txt'))      # False
print('current dir isdir?', os.path.isdir('.'))              # True

# File size
size = os.path.getsize('demo.txt')
print(f'demo.txt size: {size} bytes')

# Safe read pattern — check before opening
target = 'demo.txt'
if os.path.isfile(target):
    with open(target, 'r', encoding='utf-8') as f:
        print('Lines:', len(f.readlines()))
else:
    print(f'{target} not found')
```

```python
import os

# Practical: batch rename — add prefix to all .txt files
txt_files = [f for f in os.listdir('.') if f.endswith('.txt')]
print('Text files found:', txt_files)

for filename in txt_files:
    name, ext = os.path.splitext(filename)
    new_name = f'backup_{name}{ext}'
    os.rename(filename, new_name)
    print(f'  Renamed: {filename} → {new_name}')

# Rename back for cleanliness
for filename in os.listdir('.'):
    if filename.startswith('backup_') and filename.endswith('.txt'):
        original = filename[len('backup_'):]
        os.rename(filename, original)

print('Restored original names ✓')
```
