## 4. CSV Handling

**CSV (Comma-Separated Values)** is the most common format for tabular data. Python's built-in `csv` module handles quoting, escaping, and different delimiters automatically.

### Key classes
| Class | Purpose |
|-------|---------|
| `csv.reader` | Read rows as **lists** |
| `csv.writer` | Write rows from **lists** |
| `csv.DictReader` | Read rows as **dicts** (header → value) |
| `csv.DictWriter` | Write rows from **dicts** |

### Common parameters
| Parameter | Default | Use |
|-----------|---------|-----|
| `delimiter` | `','` | Change to `';'` or `'\t'` for TSV |
| `quotechar` | `'"'` | Character used to wrap fields containing the delimiter |
| `newline=''` | — | **Always** pass this to `open()` when using csv |

> 💡 Always open CSV files with `newline=''` to prevent double newlines on Windows.

```python
import csv

# --- csv.writer — write rows from lists ---
rows = [
    ['Name',    'Age', 'City',      'Score'],
    ['Alice',   28,    'Mumbai',    95.5],
    ['Bob',     34,    'Delhi',     87.0],
    ['Charlie', 22,    'Bangalore', 91.3],
    ['Diana',   29,    'Chennai',   78.8],
    ['Eve',     31,    'Pune',      88.6],
]

with open('students.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerows(rows)   # write all rows at once

print('students.csv written ✓')
```

```python
import csv

# --- csv.reader — read rows as lists ---
with open('students.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader)       # consume the first row as header
    print('Header:', header)
    print()
    for row in reader:
        print(row)
```

```python
import csv

# --- csv.DictReader — each row is a dict {header: value} ---
with open('students.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)   # header row used as keys automatically
    students = list(reader)

# Access by column name — much more readable!
for s in students:
    print(f"{s['Name']:<10} | Age: {s['Age']:>2} | City: {s['City']:<10} | Score: {s['Score']}")
```

```python
import csv

# --- csv.DictWriter — write dicts to CSV ---
new_students = [
    {'Name': 'Frank',  'Age': 26, 'City': 'Hyderabad', 'Score': 93.1},
    {'Name': 'Grace',  'Age': 24, 'City': 'Kolkata',   'Score': 85.5},
]

fieldnames = ['Name', 'Age', 'City', 'Score']

with open('new_students.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()         # write the header row
    writer.writerows(new_students)

# Verify
with open('new_students.csv', 'r', encoding='utf-8') as f:
    print(f.read())
```

```python
import csv

# --- Practical: calculate average score from CSV ---
with open('students.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    scores = [float(row['Score']) for row in reader]

print(f'Students   : {len(scores)}')
print(f'Average    : {sum(scores)/len(scores):.2f}')
print(f'Highest    : {max(scores)}')
print(f'Lowest     : {min(scores)}')
```
