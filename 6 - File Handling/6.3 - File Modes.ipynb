## 3. File Modes (`r`, `w`, `a`, `b`)

The **mode** argument of `open()` tells Python what you intend to do with the file.

### Text modes
| Mode | Name | File must exist? | Creates file? | Pointer starts at | Existing content |
|------|------|-----------------|--------------|-------------------|------------------|
| `'r'` | Read | ✅ Yes (error if not) | ❌ No | Beginning | Preserved |
| `'w'` | Write | ❌ No | ✅ Yes | Beginning | **Erased!** |
| `'a'` | Append | ❌ No | ✅ Yes | **End** | Preserved |
| `'x'` | Exclusive create | ❌ No (error if exists) | ✅ Yes | Beginning | N/A |
| `'r+'` | Read + Write | ✅ Yes | ❌ No | Beginning | Preserved |

### Binary modifier `b`
Append `b` to any mode for **binary** (non-text) files: `'rb'`, `'wb'`, `'ab'`

| Use binary mode for... | Example |
|------------------------|---------|
| Images | `.jpg`, `.png` |
| Audio/Video | `.mp3`, `.mp4` |
| PDFs, Excel, ZIP | `.pdf`, `.xlsx`, `.zip` |
| Exact byte copying | Any file |

> ⚠️ Using `'w'` on an existing file **silently deletes all its content**. Use `'a'` to preserve it.

```python
# Mode 'w' — creates fresh (WIPES existing content!)
with open('log.txt', 'w', encoding='utf-8') as f:
    f.write('Session started\n')
    f.write('User logged in\n')

with open('log.txt', 'r', encoding='utf-8') as f:
    print('After w:', f.read())
```

```python
# Mode 'a' — append keeps existing content
with open('log.txt', 'a', encoding='utf-8') as f:
    f.write('User performed action\n')
    f.write('Session ended\n')

with open('log.txt', 'r', encoding='utf-8') as f:
    print('After a:', f.read())
```

```python
# Mode 'x' — exclusive create (fails if file exists)
import os

try:
    with open('brand_new.txt', 'x', encoding='utf-8') as f:
        f.write('Created exclusively!\n')
    print('File created ✓')
except FileExistsError:
    print('File already exists — x mode refused to overwrite')

# Running again raises FileExistsError:
try:
    with open('brand_new.txt', 'x', encoding='utf-8') as f:
        f.write('Second attempt')
except FileExistsError as e:
    print(f'Second attempt: FileExistsError — {e}')
```

```python
# Mode 'rb' / 'wb' — binary: copy any file byte-for-byte
# Write some binary data (simulating a small file)
with open('sample.bin', 'wb') as f:
    f.write(bytes(range(256)))   # 256 bytes: 0x00 to 0xFF

# Read it back in binary mode
with open('sample.bin', 'rb') as f:
    data = f.read()

print(f'Read {len(data)} bytes')
print('First 16 bytes:', data[:16])
print('As hex:', data[:16].hex(' '))
```

```python
# Mode 'r+' — read AND write without truncating
with open('notes.txt', 'r+', encoding='utf-8') as f:
    original = f.read()
    print('Original content:')
    print(original)

    f.write('--- appended via r+ mode ---\n')  # pointer now at end

with open('notes.txt', 'r', encoding='utf-8') as f:
    print('After r+:')
    print(f.read())
```
