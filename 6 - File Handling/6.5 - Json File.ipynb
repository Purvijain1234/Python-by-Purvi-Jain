## 5. JSON Read & Write

**JSON (JavaScript Object Notation)** is the standard format for config files, APIs, and structured data. Python's built-in `json` module maps JSON types directly to Python types.

### Type mapping
| JSON | Python |
|------|--------|
| `object {}` | `dict` |
| `array []` | `list` |
| `string` | `str` |
| `number` | `int` / `float` |
| `true` / `false` | `True` / `False` |
| `null` | `None` |

### Key functions
| Function | Direction | Works with |
|----------|-----------|------------|
| `json.dumps(obj)` | Python → JSON string | any Python object |
| `json.loads(str)` | JSON string → Python | JSON string |
| `json.dump(obj, f)` | Python → JSON file | file object |
| `json.load(f)` | JSON file → Python | file object |

> 💡 Memory trick: `dumps`/`loads` have an **s** for **string**. `dump`/`load` work with **files**.

```python
import json

# Python dict → JSON string  (dumps)
person = {
    'name': 'Alice',
    'age': 28,
    'is_student': False,
    'scores': [95, 87, 92],
    'address': {'city': 'Mumbai', 'pin': '400001'},
    'nickname': None
}

json_str = json.dumps(person)                      # compact
json_pretty = json.dumps(person, indent=4)         # human-readable

print('Compact  :', json_str)
print()
print('Pretty   :')
print(json_pretty)
```

```python
import json

# JSON string → Python object  (loads)
raw = '{"name": "Bob", "age": 34, "active": true, "score": null}'

data = json.loads(raw)
print(type(data))             # <class 'dict'>
print(data['name'])           # Bob
print(data['active'])         # True  (JSON true → Python True)
print(data['score'])          # None  (JSON null → Python None)
print(type(data['age']))      # <class 'int'>
```

```python
import json

# Write Python object to a JSON FILE  (dump)
config = {
    'app_name': 'MyApp',
    'version': '2.1.0',
    'debug': False,
    'database': {
        'host': 'localhost',
        'port': 5432,
        'name': 'mydb'
    },
    'allowed_hosts': ['localhost', '127.0.0.1'],
    'max_connections': 100
}

with open('config.json', 'w', encoding='utf-8') as f:
    json.dump(config, f, indent=4)

print('config.json written ✓')
```

```python
import json

# Read JSON FILE back into Python  (load)
with open('config.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)

print('App name :', loaded['app_name'])
print('DB host  :', loaded['database']['host'])
print('DB port  :', loaded['database']['port'])
print('Hosts    :', loaded['allowed_hosts'])
print('Debug    :', loaded['debug'])    # False (bool preserved)
```

```python
import json

# Practical: update a JSON config and save it back
with open('config.json', 'r', encoding='utf-8') as f:
    cfg = json.load(f)

# Modify in-memory
cfg['version'] = '2.2.0'
cfg['debug'] = True
cfg['allowed_hosts'].append('192.168.1.1')

# Write back
with open('config.json', 'w', encoding='utf-8') as f:
    json.dump(cfg, f, indent=4)

# Confirm
with open('config.json', 'r', encoding='utf-8') as f:
    print(f.read())
```
