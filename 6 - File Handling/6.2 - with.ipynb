## 2. The `with` Statement

The `with` statement is a **context manager** that guarantees the file is closed as soon as the block finishes - even if an exception is raised.

```python
# Without with - risky
f = open('file.txt')
data = f.read()        # if this crashes, f.close() is never called!
f.close()

# With with - safe ✓
with open('file.txt') as f:
    data = f.read()    # file is ALWAYS closed when block exits
```

### Why `with` is always preferred
| | Manual `open/close` | `with` statement |
|-|--------------------|-----------------|
| Auto-close on error | ❌ No | ✅ Yes |
| Forget to close risk | ❌ High | ✅ None |
| Code clarity | ❌ Verbose | ✅ Clean |
| Resource leaks | ❌ Possible | ✅ Impossible |

> 💡 You can open **multiple files** in a single `with` statement by separating them with commas.

````python
# Basic with - read
with open('demo.txt', 'r', encoding='utf-8') as f:
    content = f.read()
# f is automatically closed here — even if an error occurred inside

print('File closed?', f.closed)   # True
print(content)
````

````python
# Basic with - write
with open('notes.txt', 'w', encoding='utf-8') as f:
    f.write('First note\n')
    f.write('Second note\n')
    f.write('Third note\n')

# Verify
with open('notes.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print(line.strip())
````

````python
# Opening multiple files in ONE with statement
with open('demo.txt', 'r', encoding='utf-8') as src, \
     open('demo_copy.txt', 'w', encoding='utf-8') as dst:
    for line in src:
        dst.write(line.upper())   # copy with uppercase transformation

# Verify the copy
with open('demo_copy.txt', 'r', encoding='utf-8') as f:
    print(f.read())
````

````python
# with handles exceptions gracefully
try:
    with open('demo.txt', 'r', encoding='utf-8') as f:
        content = f.read()
        raise ValueError('Something went wrong mid-read!')  # simulate crash
except ValueError as e:
    print('Caught error:', e)

print('File still closed after error?', f.closed)   # True — with saved us
````
