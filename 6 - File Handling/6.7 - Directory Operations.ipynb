## 7. Directory Operations

Python's `os` and `shutil` modules give you full control over **directories and the filesystem**.

### `os` — low-level directory ops
| Function | What it does |
|----------|--------------|
| `os.mkdir(path)` | Create one directory (parent must exist) |
| `os.makedirs(path, exist_ok=True)` | Create directory and all parents |
| `os.rmdir(path)` | Remove an **empty** directory |
| `os.listdir(path)` | List names in a directory |
| `os.getcwd()` | Get current working directory |
| `os.chdir(path)` | Change working directory |
| `os.rename(src, dst)` | Rename a file or directory |
| `os.walk(top)` | Recursively walk a directory tree |

### `shutil` — high-level file operations
| Function | What it does |
|----------|--------------|
| `shutil.copy(src, dst)` | Copy file (metadata not preserved) |
| `shutil.copy2(src, dst)` | Copy file + metadata |
| `shutil.copytree(src, dst)` | Copy entire directory tree |
| `shutil.rmtree(path)` | Delete directory and all its contents |
| `shutil.move(src, dst)` | Move/rename file or directory |

> ⚠️ `shutil.rmtree()` is **permanent** — there is no recycle bin. Double-check before using!

```python
import os

# Create directory structure
os.makedirs('project/src', exist_ok=True)
os.makedirs('project/data/raw', exist_ok=True)
os.makedirs('project/data/processed', exist_ok=True)
os.makedirs('project/output', exist_ok=True)

print('Created project structure ✓')

# Create some dummy files inside it
for filepath in ['project/src/main.py', 'project/data/raw/input.csv', 'project/output/report.txt']:
    with open(filepath, 'w') as f:
        f.write(f'# {filepath}\n')

print('Created dummy files ✓')
```

```python
import os

# List directory contents
print('project/ contents:', os.listdir('project'))
print('project/data/:', os.listdir('project/data'))

print()

# os.walk() — recursively walk the whole tree
for root, dirs, files in os.walk('project'):
    # Calculate depth for indentation
    depth = root.replace('project', '').count(os.sep)
    indent = '    ' * depth
    print(f'{indent}📁 {os.path.basename(root)}/')
    for file in files:
        print(f'{indent}    📄 {file}')
```

```python
import shutil, os

# shutil.copy2 — copy a file preserving metadata
shutil.copy2('project/data/raw/input.csv', 'project/data/processed/input_clean.csv')
print('Copied input.csv → input_clean.csv ✓')

# shutil.move — move a file
shutil.move('project/output/report.txt', 'project/report_final.txt')
print('Moved report.txt → report_final.txt ✓')

# shutil.copytree — copy an entire directory tree
shutil.copytree('project/data', 'project/data_backup')
print('Copied data/ → data_backup/ ✓')

print('\nproject/ after operations:')
print(os.listdir('project'))
```

```python
import os

# os.walk() — find all Python files in a directory tree
print('All .py files under project/:')
py_files = []

for root, dirs, files in os.walk('project'):
    for file in files:
        if file.endswith('.py'):
            full_path = os.path.join(root, file)
            size = os.path.getsize(full_path)
            py_files.append(full_path)
            print(f'  {full_path}  ({size} bytes)')

print(f'\nTotal .py files: {len(py_files)}')
```

```python
import shutil, os

# shutil.rmtree — delete an entire directory tree (permanent!)
shutil.rmtree('project')
print('project/ deleted ✓')
print('Exists?', os.path.exists('project'))  # False

# Clean up all other temp files created in this notebook
for f in ['demo.txt', 'demo_copy.txt', 'fruits.txt', 'notes.txt',
          'log.txt', 'brand_new.txt', 'sample.bin',
          'students.csv', 'new_students.csv', 'config.json']:
    if os.path.exists(f):
        os.remove(f)
        print(f'  Removed {f}')

print('\nAll cleanup done ✓')
```
