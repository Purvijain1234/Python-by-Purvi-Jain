## 1. `open()`, read, write

`open()` is Python's built-in function for working with files. It returns a **file object** that you can read from or write to.

### Signature
```python
open(file, mode='r', encoding=None)
```

### Reading methods on a file object
| Method | Returns | When to use |
|--------|---------|-------------|
| `.read()` | Entire file as one string | Small files |
| `.readline()` | One line at a time | Streaming large files |
| `.readlines()` | List of all lines | When you need indexing |
| Iterate `for line in f` | One line per iteration | Most efficient |

### Writing methods
| Method | What it does |
|--------|--------------|
| `.write(str)` | Writes a string (no automatic newline) |
| `.writelines(list)` | Writes a list of strings (no automatic newlines) |

> ⚠️ Always call `.close()` after you finish — or better yet, use the `with` statement (Section 2) which closes automatically.

````python
# --- Writing to a file ---
f = open('demo.txt', 'w', encoding='utf-8')   # 'w' creates or overwrites
f.write('Line 1: Hello, File Handling!\n')
f.write('Line 2: Python makes I/O easy.\n')
f.write('Line 3: Always close your files.\n')
f.close()                                      # must close manually

print('demo.txt written ✓')
````

````python
# --- read() - entire file as one string ---
f = open('demo.txt', 'r', encoding='utf-8')
content = f.read()
f.close()

print(repr(content))   # shows \n characters explicitly
print()
print(content)         # printed as formatted text
````

````python
# --- readline() — one line at a time ---
f = open('demo.txt', 'r', encoding='utf-8')

first  = f.readline()   # reads up to and including '\n'
second = f.readline()
f.close()

print('First line  :', first.strip())
print('Second line :', second.strip())
````

````python
# --- readlines() — list of all lines ---
f = open('demo.txt', 'r', encoding='utf-8')
lines = f.readlines()   # ['Line 1: ...\n', 'Line 2: ...\n', ...]
f.close()

print(f'Total lines: {len(lines)}')
for i, line in enumerate(lines, start=1):
    print(f'  [{i}] {line.rstrip()}')
````

````python
# --- writelines() — write a list of strings ---
f = open('fruits.txt', 'w', encoding='utf-8')
fruits = ['apple\n', 'banana\n', 'cherry\n', 'mango\n']
f.writelines(fruits)   # note: you must add '\n' yourself
f.close()

# Verify by iterating directly over the file object
f = open('fruits.txt', 'r', encoding='utf-8')
for line in f:         # most memory-efficient way to iterate
    print(line.strip())
f.close()
````
