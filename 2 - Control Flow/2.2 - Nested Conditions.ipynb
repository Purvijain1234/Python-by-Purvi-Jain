## 2. Nested Conditions

You can place an `if` statement **inside** another `if` block. Each level of nesting requires an additional 4 spaces of indentation.

### When to use nested vs combined conditions
| Situation | Prefer |
|-----------|--------|
| Both checks are independent | `if A and B:` (single condition) |
| Second check only makes sense after the first passes | nested `if` |
| Three or more levels deep | refactor into a function |

> 💡 Deep nesting (3+ levels) is a code smell. Flatten with **guard clauses** - check the failure condition first and `return` / `continue` early.

```python
# Nested if — login system example
is_registered = True
password = "python123"
is_admin = False

if is_registered:
    if password == "python123":
        print("Login successful!")
        if is_admin:
            print("Admin dashboard loaded.")
        else:
            print("User dashboard loaded.")
    else:
        print("Wrong password.")
else:
    print("Please register first.")
```

```python
# ---- Guard clause pattern (flattened, easier to read) ----
def process_order(item, stock, payment_ok):
    if not item:
        print("No item selected.")      # exit early
        return
    if stock <= 0:
        print("Out of stock.")          # exit early
        return
    if not payment_ok:
        print("Payment failed.")        # exit early
        return
    # happy path — only reached when all checks pass
    print(f"Order for '{item}' placed successfully!")

process_order("Laptop", 5, True)
process_order("Phone",  0, True)
process_order("",       5, True)
```
