## 1. if / elif / else

The `if` statement evaluates a **condition** (any expression that is truthy or falsy) and runs a block of code only when that condition is `True`.

```
if condition_1:
    # runs when condition_1 is True
elif condition_2:
    # runs when condition_1 is False AND condition_2 is True
else:
    # runs when ALL conditions above are False
```

### Rules
- You can have **zero or more** `elif` blocks.
- The `else` block is optional and always comes last.
- Python checks conditions **top to bottom** and stops at the first `True` one.
- Indentation (4 spaces) defines the block - not curly braces.

### One-liner (Ternary operator)
```python
value_if_true if condition else value_if_false
```

````python
# Basic if / elif / else
marks = 72

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "F"

print(f"Marks: {marks} → Grade: {grade}")
````

````python
# Truthy / falsy conditions (not just comparisons)
username = ""
if username:
    print(f"Welcome, {username}!")
else:
    print("No username provided.")

print()
````

````python
# Combining conditions with 'and' / 'or'
age = 20
has_id = True

if age >= 18 and has_id:
    print("Entry allowed.")
else:
    print("Entry denied.")
````

````python
# Ternary (one-liner)
num = 7
parity = "even" if num % 2 == 0 else "odd"
print(f"{num} is {parity}.")
````

````python
# Chained comparisons (Pythonic!)
temperature = 22
if 18 <= temperature <= 26:
    print("Comfortable temperature.")
````
