## 4. while Loops

A `while` loop keeps running **as long as its condition is `True`**. Use it when you don't know in advance how many iterations you need.

```python
while condition:
    # body
```

### for vs while — when to use which
| Scenario | Use |
|----------|-----|
| Known number of iterations / iterating a collection | `for` |
| Repeat until a condition changes (user input, game loop, retry logic) | `while` |

> Always make sure something inside the loop **eventually makes the condition False**, otherwise you get an **infinite loop**. Press `Kernel → Interrupt` in Jupyter if you accidentally run one.

````python
# Basic while - countdown
count = 5
while count > 0:
    print(count, end=" ")
    count -= 1          # IMPORTANT: moves toward the exit condition
print("Go!")
````

````python
# while with a flag variable
attempts = 0
max_attempts = 3
correct_pin = "1234"

# Simulated PIN entries instead of real input()
test_pins = ["0000", "9999", "1234"]

while attempts < max_attempts:
    entered = test_pins[attempts]          # simulate user input
    attempts += 1
    print(f"Attempt {attempts}: entered '{entered}'")
    if entered == correct_pin:
        print("Access granted!")
        break
else:
    # while-else: runs only if loop finished WITHOUT hitting break
    print("Too many failed attempts. Account locked.")
````

````python
# Accumulator pattern - sum until threshold
total = 0
number = 1
while total < 20:
    total += number
    number += 1
print(f"Reached total {total} after adding numbers 1 to {number - 1}")
````
