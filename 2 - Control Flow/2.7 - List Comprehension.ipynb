## 7. List Comprehensions

A **list comprehension** is a compact, readable way to create a new list by transforming or filtering an existing iterable — all in one line.

```python
# Full syntax
[expression  for item in iterable  if condition]
#  ^^what       ^^loop              ^^optional filter
```

### Comprehension types
| Syntax | Creates |
|--------|--------|
| `[expr for x in it]` | **list** |
| `{expr for x in it}` | **set** (no duplicates) |
| `{k: v for x in it}` | **dict** |
| `(expr for x in it)` | **generator** (lazy, memory-efficient) |

### When NOT to use comprehensions
- The expression is complex enough to need comments → use a regular loop.
- You have side effects (printing, appending to external list) → use a regular loop.
- Nested more than 2 levels deep → use a regular loop.

## List Comprehension

````python
# Basic: squares of 1-10
squares = [x**2 for x in range(1, 11)]
print("Squares:", squares)
````

````python
# With filter: even numbers only
evens = [x for x in range(1, 21) if x % 2 == 0]
print("Evens:", evens)
````

````python
# Transform: uppercase all strings
fruits = ["apple", "banana", "mango"]
upper_fruits = [f.upper() for f in fruits]
print("Upper:", upper_fruits)
````

````python
# With condition on expression (if-else BEFORE for)
labels = ["even" if x % 2 == 0 else "odd" for x in range(1, 7)]
print("Labels:", labels)
````

## Set Comprehension

````python
sentence = "the cat sat on the mat"
unique_words = {word for word in sentence.split()}
print("Unique words:", unique_words)   # order not guaranteed
````

## Dict Comprehension

````python
names  = ["Alice", "Bob", "Charlie"]
scores = [88, 95, 72]
gradebook = {name: score for name, score in zip(names, scores)}
print("Gradebook:", gradebook)
````

````python
# Invert a dictionary (swap keys and values)
inverted = {v: k for k, v in gradebook.items()}
print("Inverted:", inverted)
````

## Nested List Comprehension

````python
# Flatten a 2D matrix into 1D
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print("Flat:", flat)
````

````python
# --- Generator Expression (memory-efficient) ---
# sum of squares from 1 to 1,000,000 without building a list in memory
total = sum(x**2 for x in range(1, 1_000_001))
print(f"Sum of squares 1..1,000,000 = {total:,}")
````
