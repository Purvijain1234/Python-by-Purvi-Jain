## 6. range() Usage

`range()` generates a sequence of integers **on demand** (lazy - does not store all values in memory). It's most commonly used with `for` loops.

### Signatures
```python
range(stop)              # 0, 1, 2, ..., stop-1
range(start, stop)       # start, start+1, ..., stop-1
range(start, stop, step) # start, start+step, ..., < stop
```

### Key behaviours
- `stop` is **exclusive** - `range(5)` gives 0 through 4, never 5.
- `step` can be **negative** to count downward.
- `range` supports `len()`, indexing, and `in` checks.
- Convert to a list with `list(range(...))` if you need all values at once.

````python
# range(stop) — 0 to n-1
print(list(range(5)))           # [0, 1, 2, 3, 4]
````

````python
# range(start, stop)
print(list(range(1, 6)))        # [1, 2, 3, 4, 5]
````

````python
# range(start, stop, step)
print(list(range(0, 20, 5)))    # [0, 5, 10, 15]
````

````python
# Counting DOWN with negative step
print(list(range(10, 0, -2)))   # [10, 8, 6, 4, 2]
````

````python
# Practical - print squares 1² to 10²
print("Squares:")
for n in range(1, 11):
    print(f"  {n:2d}² = {n**2:3d}")

print()
````

````python
# range supports indexing and membership checks
r = range(0, 100, 10)
print(r[3])          # 30
print(50 in r)       # True
print(55 in r)       # False
print(len(r))        # 10
````

````python
# Reverse a list using range
items = ["a", "b", "c", "d"]
for i in range(len(items) - 1, -1, -1):
    print(items[i], end=" ")
print()
````

````python
# More Pythonic way - reversed()
for item in reversed(items):
    print(item, end=" ")
print()
````
