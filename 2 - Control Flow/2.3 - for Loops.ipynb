## 3. for Loops

A `for` loop **iterates over any iterable** - a list, string, tuple, dictionary, range, file, and more. On each iteration, the loop variable takes the next value from the sequence.

```python
for variable in iterable:
    # body runs once per item
```

### Useful built-ins to use with for loops
| Function | What it does |
|----------|--------------|
| `range(n)` | integers 0 to n-1 |
| `enumerate(seq)` | pairs of (index, value) |
| `zip(a, b)` | pairs items from two iterables |
| `reversed(seq)` | iterates in reverse |
| `sorted(seq)` | iterates in sorted order |

````python
# Iterating over a list
fruits = ["apple", "banana", "mango", "cherry"]
for fruit in fruits:
    print(fruit)
````

````python
# Iterating over a string (character by character)
for char in "Python":
    print(char, end=" ")
print()
````

````python
# enumerate — when you need both index AND value
students = ["Alice", "Bob", "Charlie"]
for i, student in enumerate(students, start=1):   # start=1 → 1-based numbering
    print(f"{i}. {student}")
````

````python
# zip — loop over two lists in parallel
names  = ["Alice", "Bob", "Charlie"]
scores = [88, 95, 72]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

````python
# Iterating over a dictionary
student = {"name": "Purvi", "age": 21, "grade": "A"}
for key, value in student.items():
    print(f"{key}: {value}")
````

````python
# Nested for loop - multiplication table
for row in range(1, 4):
    for col in range(1, 4):
        print(f"{row}x{col}={row*col}", end="  ")
    print()
````

````python
# Nested for loop — multiplication table
for row in range(1, 4):
    for col in range(1, 4):
        print("*", end="  ")
    print()
````
