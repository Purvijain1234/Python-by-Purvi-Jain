## 5. break / continue / pass

These three keywords give you fine-grained control over what happens inside a loop (or block).

| Keyword | Effect |
|---------|--------|
| `break` | **Exit the loop immediately** - skips the rest of the loop body AND any `else` clause |
| `continue` | **Skip the rest of this iteration** - jump back to the top of the loop for the next item |
| `pass` | **Do nothing** - a no-op placeholder used where Python syntax requires a statement but you have nothing to put yet |

### The loop `else` clause (often overlooked!)
Both `for` and `while` loops can have an `else` block.  
- It runs when the loop **completes normally** (no `break`).
- It does **not** run if the loop was terminated by `break`.
- Most useful for **search patterns**: did we find the item or not?

```python
# break - stop as soon as we find the target
numbers = [3, 7, 2, 9, 4, 6, 1]
target = 9

for i, n in enumerate(numbers):
    if n == target:
        print(f"Found {target} at index {i}")
        break
else:
    print(f"{target} not found in the list.")
```

```python
# continue - skip unwanted items
print("Even numbers only:")
for n in range(1, 11):
    if n % 2 != 0:
        continue        # skip odd numbers
    print(n, end=" ")
print()
```

```python
# continue - skip None / empty values
data = ["Alice", None, "Bob", "", "Charlie", None]
clean = []
for item in data:
    if not item:
        continue
    clean.append(item)
print("Cleaned list:", clean)
```

```python
# pass - placeholder for code you'll write later
for n in range(5):
    if n == 3:
        pass            # TODO: handle n == 3 later
    else:
        print(n, end=" ")
print()
```

```python
# Nested loops - break only exits the INNERMOST loop
print("Finding first pair that sums to 10:")
found = False
for a in range(1, 8):
    for b in range(1, 8):
        if a + b == 10:
            print(f"  {a} + {b} = 10")
            found = True
            break       # exits inner loop only
    if found:
        break           # exits outer loop
```
