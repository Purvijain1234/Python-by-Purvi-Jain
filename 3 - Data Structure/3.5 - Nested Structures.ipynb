## 5. Nested Structures

Real-world data is almost never flat. Python lets you nest any data structure inside any other - lists of dicts, dicts of lists, dicts of dicts, etc.

### Common patterns
| Pattern | Example use-case |
|---------|------------------|
| List of dicts | Database rows, JSON API responses, CSV records |
| Dict of lists | Grouping items by category |
| Dict of dicts | Config files, nested objects |
| List of lists | Matrix, grid, 2-D table |

> 💡 Use `json.dumps(data, indent=2)` to pretty-print any nested structure - great for debugging.

```python
import json

# --- List of dicts (like database rows) ---
students = [
    {"name": "Alice",   "grade": "A", "scores": [88, 92, 95]},
    {"name": "Bob",     "grade": "B", "scores": [75, 80, 70]},
    {"name": "Charlie", "grade": "C", "scores": [60, 65, 58]},
]

# Access a specific field
print(students[0]["name"])            # Alice
print(students[1]["scores"][2])       # 70  (Bob's 3rd score)

# Iterate and compute
print("\nStudent averages:")
for s in students:
    avg = sum(s["scores"]) / len(s["scores"])
    print(f"  {s['name']}: {avg:.1f}")

# Sort list of dicts by a field
by_name = sorted(students, key=lambda s: s["name"])
print("\nSorted by name:", [s["name"] for s in by_name])
```

```python
# --- Dict of dicts (nested config / profile) ---
config = {
    "database": {
        "host": "localhost",
        "port": 5432,
        "name": "mydb"
    },
    "server": {
        "host": "0.0.0.0",
        "port": 8000,
        "debug": True
    }
}

print("DB host:", config["database"]["host"])
print("Server port:", config["server"]["port"])

# Safe nested access with .get()
timeout = config.get("database", {}).get("timeout", 30)
print("DB timeout (default):", timeout)
```

```python
# --- 2-D list (matrix) ---
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Matrix:")
for row in matrix:
    print(" ", row)

print("Element [1][2]:", matrix[1][2])  # 6

# Transpose using list comprehension
transposed = [[matrix[r][c] for r in range(3)] for c in range(3)]
print("Transposed:")
for row in transposed:
    print(" ", row)
```

```python
# Pretty-print nested structure
print(json.dumps(config, indent=2))
```
