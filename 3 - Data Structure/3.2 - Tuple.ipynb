## 2. Tuples

A **tuple** is an **ordered, immutable** sequence. Once created, its items cannot be changed, added, or removed. This makes tuples:
- **Faster** than lists for iteration and access.
- **Safer** - they signal to the reader that this data should not change.
- **Hashable** (if all items are hashable) - so they can be used as **dictionary keys** or **set members**.

### When to use a tuple vs a list
| Use a **tuple** when… | Use a **list** when… |
|----------------------|---------------------|
| Data won't change (coordinates, RGB colour, DB row) | Data needs to grow/shrink |
| Returning multiple values from a function | Building a collection dynamically |
| Using as a dictionary key | Modifying items individually |

> A single-item tuple needs a trailing comma: `(42,)`- without it `(42)` is just the integer `42` in parentheses.

```python
# Creating tuples
empty_t  = ()
single   = (42,)              # trailing comma is REQUIRED for single-item
point    = (3, 7)
rgb      = (255, 128, 0)
mixed_t  = ("Alice", 21, True)
```

```python
print(type(single), single)   # <class 'tuple'> (42,)
print(type((42)),   (42))     # <class 'int'>   42  ← NOT a tuple!
```

```python
# Indexing & slicing (same as lists)
print(rgb[0])         # 255
print(rgb[-1])        # 0
print(rgb[0:2])       # (255, 128)
```

```python
# Immutability - this will raise TypeError
try:
    rgb[0] = 100
except TypeError as e:
    print("TypeError:", e)
```

```python
# Tuple unpacking
x, y = point
print(f"x={x}, y={y}")

name, age, active = mixed_t
print(f"name={name}, age={age}, active={active}")
```

```python
# Swap two variables (uses tuple packing/unpacking under the hood)
a, b = 10, 20
a, b = b, a
print(f"After swap: a={a}, b={b}")
```

```python
# Returning multiple values from a function
def min_max(numbers):
    return min(numbers), max(numbers)   # returns a tuple

lo, hi = min_max([4, 1, 9, 2, 7])
print(f"min={lo}, max={hi}")
```

```python
# Tuple as dict key (lists cannot be dict keys)
locations = {
    (28.6, 77.2): "New Delhi",
    (19.0, 72.8): "Mumbai",
    (13.0, 80.2): "Chennai",
}
print(locations[(28.6, 77.2)])
```

```python
# Useful tuple methods
t = (1, 2, 3, 2, 4, 2)
print("count(2):", t.count(2))   # 3
print("index(3):", t.index(3))   # 2
```
