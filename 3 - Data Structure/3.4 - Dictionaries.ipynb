## 4. Dictionaries

A **dictionary** stores data as **key-value pairs**. As of Python 3.7+ dictionaries maintain **insertion order**. Keys must be unique and hashable (strings, numbers, tuples); values can be anything.

### Key methods
| Method | What it does |
|--------|--------------|
| `d[key]` | Get value - raises `KeyError` if missing |
| `d.get(key, default)` | Get value safely - returns `default` if missing |
| `d[key] = val` | Set / update a value |
| `d.keys()` | View of all keys |
| `d.values()` | View of all values |
| `d.items()` | View of all (key, value) pairs |
| `d.update(other)` | Merge another dict in |
| `d.pop(key)` | Remove and return value |
| `d.setdefault(key, val)` | Return value; set it to `val` if key missing |
| `key in d` | Membership check on keys |
| `len(d)` | Number of key-value pairs |

```python
# Creating dictionaries
empty_d = {}
student = {
    "name":   "Purvi",
    "age":    21,
    "grade":  "A",
    "active": True
}

# Also via dict()
person = dict(name="Alice", age=30, city="Delhi")
print(student)
print(person)
```

```python
# Accessing values
print(student["name"])               # direct — KeyError if missing
print(student.get("score", "N/A"))   # safe — returns default

# Adding / updating
student["score"] = 95                # add new key
student["age"]   = 22                # update existing key
print(student)
```

```python
# Iterating
for key in student.keys():
    print(key, end=" ")
print()

for value in student.values():
    print(value, end=" ")
print()

for k, v in student.items():
    print(f"  {k}: {v}")
```

```python
# pop and update
removed = student.pop("active")
print(f"Popped: {removed}")

extra = {"email": "purvi@example.com", "city": "Delhi"}
student.update(extra)
print("After update:", student)
```

```python
# setdefault — useful for grouping / counting patterns
word_count = {}
sentence = "the cat sat on the mat the cat"
for word in sentence.split():
    word_count[word] = word_count.get(word, 0) + 1
print("Word count:", word_count)
```

```python
# Grouping with setdefault
people = [
    {"name": "Alice", "dept": "Engineering"},
    {"name": "Bob",   "dept": "Marketing"},
    {"name": "Carol", "dept": "Engineering"},
    {"name": "Dave",  "dept": "Marketing"},
]
by_dept = {}
for p in people:
    by_dept.setdefault(p["dept"], []).append(p["name"])
print("By department:", by_dept)
```
