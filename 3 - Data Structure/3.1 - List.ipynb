## 1. Lists & Indexing

A **list** is an **ordered, mutable** collection that can hold items of any type (even mixed types). It is the most versatile and commonly used data structure in Python.

### Indexing
- **Positive index** starts at `0` from the left.
- **Negative index** starts at `-1` from the right.

```
 List:   [ 'a',  'b',  'c',  'd',  'e' ]
 Pos:       0     1     2     3     4
 Neg:      -5    -4    -3    -2    -1
```

### Key list methods
| Method | What it does |
|--------|--------------|
| `.append(x)` | Add `x` to the end |
| `.insert(i, x)` | Insert `x` at index `i` |
| `.extend(iterable)` | Add all items from iterable |
| `.remove(x)` | Remove first occurrence of `x` |
| `.pop(i)` | Remove & return item at index `i` (default: last) |
| `.index(x)` | Return index of first `x` |
| `.count(x)` | Count occurrences of `x` |
| `.sort()` | Sort in-place |
| `.reverse()` | Reverse in-place |
| `.copy()` | Shallow copy |
| `.clear()` | Remove all items |

````python
# Creating lists
empty   = []
numbers = [10, 20, 30, 40, 50]
mixed   = [1, "hello", 3.14, True, None]

print("Numbers:", numbers)
print("Mixed:  ", mixed)
````

````python
# Indexing
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[0])    # apple      (first)
print(fruits[-1])   # elderberry (last)
print(fruits[-2])   # date       (second from end)
````

````python
# Modifying
fruits[1] = "blueberry"          # replace by index
print("After replace:", fruits)
````

````python
fruits.append("fig")             # add to end
fruits.insert(2, "coconut")      # insert at index 2
print("After append+insert:", fruits)
````

````python
fruits.remove("date")            # remove by value
popped = fruits.pop()            # remove & return last item
print(f"Removed 'date', popped '{popped}':", fruits)
````

````python
# Searching
nums = [3, 1, 4, 1, 5, 9, 2, 6, 1]
print("Index of first 1:", nums.index(1))
print("Count of 1:",        nums.count(1))
print("Is 5 in list:",      5 in nums)
````

````python
# Sorting
nums_copy = nums.copy()
nums_copy.sort()                   # in-place, modifies original
print("Sorted (asc):", nums_copy)
````

````python
sorted_desc = sorted(nums, reverse=True)  # returns new list, original unchanged
print("Sorted desc: ", sorted_desc)
print("Original:    ", nums)
````

````python
# Combining lists
a = [1, 2, 3]
b = [4, 5, 6]
print("Concatenate:  ", a + b)       # new list
a.extend(b)                          # modifies a in-place
print("After extend: ", a)
print("Repeat:       ", [0] * 5)
````
