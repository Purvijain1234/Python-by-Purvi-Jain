## 3. Sets

A **set** is an **unordered collection of unique items**. Duplicate values are automatically discarded. Sets are implemented as hash tables, which makes membership testing (`in`) extremely fast -- O(1) on average.

### When to use a set
- Remove duplicates from a list.
- Fast membership tests on large collections.
- Set mathematics: union, intersection, difference.

### Set operations
| Operation | Operator | Method |
|-----------|----------|--------|
| Union (all items) | `A \| B` | `A.union(B)` |
| Intersection (common) | `A & B` | `A.intersection(B)` |
| Difference (in A not B) | `A - B` | `A.difference(B)` |
| Symmetric difference (in one but not both) | `A ^ B` | `A.symmetric_difference(B)` |
| Subset | `A <= B` | `A.issubset(B)` |
| Superset | `A >= B` | `A.issuperset(B)` |

> ⚠️ `{}` creates an **empty dict**, NOT a set. Use `set()` for an empty set.

```python
# Creating sets
empty_set = set()                      # NOT {} which is an empty dict
colours   = {"red", "green", "blue"}
from_list = set([1, 2, 2, 3, 3, 3])   # duplicates removed

print(colours)
print(from_list)   # {1, 2, 3}
```

```python
# Deduplication use-case
votes = ["Alice", "Bob", "Alice", "Charlie", "Bob", "Alice"]
unique_voters = set(votes)
print("Unique voters:", unique_voters)
print("Total unique:", len(unique_voters))
```

```python
# Membership - O(1) average (much faster than list for large data)
banned_words = {"spam", "scam", "fake", "free", "click"}
msg = "Get free money now"
flagged = [w for w in msg.lower().split() if w in banned_words]
print("Flagged words:", flagged)
```

```python
# Mutating sets
s = {1, 2, 3}
s.add(4)              # add one item
s.update([5, 6, 7])   # add multiple items
print("After add+update:", s)

s.discard(10)         # safe remove — no error if not found
s.remove(1)           # raises KeyError if not found
print("After discard+remove:", s)
```

```python
# Set operations
python_students = {"Alice", "Bob", "Charlie", "Diana"}
java_students   = {"Bob", "Diana", "Eve", "Frank"}

print("Union (all):",         python_students | java_students)
print("Intersection (both):", python_students & java_students)
print("Python only:",         python_students - java_students)
print("Java only:",           java_students - python_students)
print("In exactly one:",      python_students ^ java_students)
```

```python

# Subset / Superset
A = {1, 2}
B = {1, 2, 3, 4}
print(f"A ⊆ B: {A <= B}")   # True
print(f"B ⊇ A: {B >= A}")   # True
print(f"A disjoint from {{5,6}}: {A.isdisjoint({5, 6})}")  # True
```
