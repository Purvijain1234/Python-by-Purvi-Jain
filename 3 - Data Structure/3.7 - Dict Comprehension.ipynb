## 7. Dict / Set Comprehensions

Just like list comprehensions, Python supports concise one-line construction for **dicts** and **sets**.

```python
# Dict comprehension
{key_expr: value_expr  for item in iterable  if condition}

# Set comprehension
{expr  for item in iterable  if condition}
```

### Common use-cases
| Task | Comprehension |
|------|---------------|
| Invert a dict | `{v: k for k, v in d.items()}` |
| Filter dict by value | `{k: v for k, v in d.items() if v > 0}` |
| Square mapping | `{n: n**2 for n in range(1, 6)}` |
| Unique first letters | `{word[0] for word in words}` |

## Dict Comprehension

````python
# Basic: number → square mapping
squares = {n: n**2 for n in range(1, 8)}
print("Squares dict:", squares)
````

````python
# From two lists using zip
keys   = ["name", "age", "city"]
values = ["Purvi", 21, "Delhi"]
profile = {k: v for k, v in zip(keys, values)}
print("Profile:", profile)
````

````python
# Filter: keep only passing scores (>= 60)
scores = {"Alice": 88, "Bob": 45, "Charlie": 72, "Diana": 55, "Eve": 91}
passing = {name: score for name, score in scores.items() if score >= 60}
print("Passing:", passing)
````

````python
# Transform values: convert Celsius to Fahrenheit
celsius    = {"freezing": 0, "body": 37, "boiling": 100}
fahrenheit = {k: round(v * 9/5 + 32, 1) for k, v in celsius.items()}
print("Fahrenheit:", fahrenheit)
````

````python
# Invert a dictionary
country_code = {"India": "IN", "USA": "US", "Germany": "DE"}
code_country = {v: k for k, v in country_code.items()}
print("Inverted:", code_country)
````

````python
# Nested dict comprehension: multiply each value in sub-dict
prices = {"fruits": {"apple": 50, "banana": 20}, "vegs": {"tomato": 30}}
discounted = {cat: {item: round(p * 0.9) for item, p in items.items()}
              for cat, items in prices.items()}
print("Discounted:", discounted)
````

## Set Comprehension

````python
# Unique lengths of words
words = ["hi", "hello", "hey", "world", "python", "code"]
unique_lengths = {len(w) for w in words}
print("Unique lengths:", unique_lengths)
````

````python
# First letters
first_letters = {w[0].upper() for w in words}
print("First letters:", first_letters)
````

````python
# Common items between two lists (using set comprehension + intersection)
list_a = [1, 2, 3, 4, 5, 6]
list_b = [4, 5, 6, 7, 8, 9]
common = {x for x in list_a if x in list_b}
print("Common:", common)    # or simply: set(list_a) & set(list_b)
````
