## 1. `import` & `from import`

A **module** is any `.py` file containing Python code - functions, classes, and variables that can be reused across programs. Python's module system is one of its greatest strengths.

### Ways to import
| Syntax | Usage | Access style |
|--------|-------|-------------|
| `import module` | Entire module | `module.name` |
| `import module as alias` | Entire module with shorthand | `alias.name` |
| `from module import name` | Specific item only | `name` directly |
| `from module import name as alias` | Specific item with shorthand | `alias` directly |
| `from module import *` | Everything (⚠️ avoid) | `name` directly |

> 💡 Always prefer **explicit** imports - they make code easier to read and debug.

```python
# import - access everything via dot notation
import math

print(math.pi)           # 3.141592653589793
print(math.sqrt(16))     # 4.0
print(math.factorial(5)) # 120
```

```python
# import as - give the module a shorter alias
import math as m

print(m.ceil(4.2))    # 5
print(m.floor(4.9))   # 4
print(m.log(100, 10)) # 2.0
```

```python
# from import - bring specific names into scope
from math import pi, sqrt, factorial

print(pi)            # 3.141592653589793 — no 'math.' prefix needed
print(sqrt(25))      # 5.0
print(factorial(6))  # 720
```

```python
from math import sqrt as square_root

print(square_root(144))  # 12.0
```

```python
# Inspect what's available inside a module
import math
print([name for name in dir(math) if not name.startswith('_')])
```

```python
# Importing from nested packages (package.subpackage.module)
from collections import OrderedDict, defaultdict
```

```python
# defaultdict never raises KeyError — provides a default value
word_count = defaultdict(int)
for word in ["apple", "banana", "apple", "cherry", "banana", "apple"]:
    word_count[word] += 1

print(dict(word_count))  # {'apple': 3, 'banana': 2, 'cherry': 1}
```
