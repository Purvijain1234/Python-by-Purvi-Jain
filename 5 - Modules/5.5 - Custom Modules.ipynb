## 5. Creating Custom Modules

A **module** is simply a `.py` file. A **package** is a folder containing an `__init__.py` file.

### Project layout example
```
my_project/
├── main.py              ← entry point
├── utils.py             ← module: utility helpers
├── calculator.py        ← module: math logic
└── mypackage/           ← package (directory)
    ├── __init__.py      ← marks it as a package; can re-export names
    ├── greetings.py     ← sub-module
    └── validators.py    ← sub-module
```

### Key module attributes
| Attribute | What it holds |
|-----------|---------------|
| `__name__` | Module name (or `'__main__'` when run directly) |
| `__doc__` | Module docstring |
| `__file__` | Full path to the `.py` file |
| `__all__` | List of public names exported by `from mod import *` |

> 💡 Use a leading underscore (`_helper`) to signal that a function is **private** / internal.

````python
# Creating a custom module file

module_code = """
def greet(name):
    return f"Hello, {name}! Welcome to Python."

def add(a, b):
    return a + b

def square(n):
    return n * n
"""

# Save as mymodule.py
with open("mymodule.py", "w") as f:
    f.write(module_code)

print("Custom module 'mymodule.py' created successfully!")
````

````python
# Step 2 — Import and use the custom module
import mymodule

# Using functions from module
print(mymodule.greet("Purvi"))
print("Addition:", mymodule.add(5, 3))
print("Square:", mymodule.square(4))
````

````python
# Step 3 - Import Specific Functions
from mymodule import add, square

print("Add:", add(10, 20))
print("Square:", square(6))
````
