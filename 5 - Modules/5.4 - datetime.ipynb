## 4. `datetime`

The `datetime` module handles **dates, times, and time arithmetic**.

### Core Classes
| Class | Stores | Example |
|-------|--------|---------|
| `date` | Year, month, day | `date(2024, 3, 15)` |
| `time` | Hour, minute, second, microsecond | `time(9, 30, 0)` |
| `datetime` | Date + Time combined | `datetime(2024, 3, 15, 9, 30)` |
| `timedelta` | Duration / interval | `timedelta(days=7)` |

### Format Codes for `strftime` / `strptime`
| Code | Meaning | Example |
|------|---------|--------|
| `%Y` | 4-digit year | `2024` |
| `%m` | Month `01–12` | `03` |
| `%d` | Day `01–31` | `15` |
| `%H` | Hour 24h `00–23` | `14` |
| `%M` | Minute `00–59` | `30` |
| `%A` | Full weekday name | `Friday` |
| `%B` | Full month name | `March` |
| `%I:%M %p` | 12-hour clock + AM/PM | `02:30 PM` |

> 💡 `strftime` → datetime **to** string. `strptime` → string **to** datetime.

```python
from datetime import date, time, datetime, timedelta
```

```python
# --- date ---
birthday = date(2007, 8, 27)
today = date.today()
print("Birthday:", birthday)
print("Today:", today)
print("Weekday (0=Mon):", today.weekday())
```

```python
# --- time ---
alarm = time(7, 30, 0)
print("\nAlarm:", alarm)      # 07:30:00
```

```python
# --- datetime ---
now = datetime.now()
launch = datetime(2024, 3, 15, 9, 30, 0)
print("\nNow:", now)
print("Launch:", launch)
print("Date part:", now.date())
print("Time part:", now.time())
```

```python
now = datetime.now()
```

```python
# Arithmetic with timedelta
tomorrow     = now + timedelta(days=1)
next_week    = now + timedelta(weeks=1)
in_90_mins   = now + timedelta(minutes=90)
three_months = now + timedelta(days=90)

print(f"Tomorrow     : {tomorrow.date()}")
print(f"Next week    : {next_week.date()}")
print(f"In 90 minutes: {in_90_mins.strftime('%H:%M')}")
print(f"In 90 days   : {three_months.date()}")
```

```python
# Difference between two datetimes
start = datetime(2024, 1, 1)
end   = datetime(2024, 12, 31)
diff  = end - start
print(f"\nDays in 2024 : {diff.days}")
print(f"In seconds   : {diff.total_seconds():,.0f}")
```

```python
# strftime — datetime → formatted string
print(now.strftime('%Y-%m-%d'))               # 2024-03-15
print(now.strftime('%d/%m/%Y'))               # 15/03/2024
print(now.strftime('%B %d, %Y'))              # March 15, 2024
print(now.strftime('%A, %d %B %Y'))           # Friday, 15 March 2024
print(now.strftime('%I:%M %p'))               # 09:30 AM
print(now.strftime('%Y-%m-%d %H:%M:%S'))      # 2024-03-15 09:30:00
```

```python
# strptime — string → datetime
date_str = 'March 15, 2024'
dt = datetime.strptime(date_str, '%B %d, %Y')
print(f"\nParsed: {dt}  | year={dt.year}, month={dt.month}")
```

```python
# Unix timestamp
ts = now.timestamp()
print(f"Unix timestamp : {ts:.0f}")
print(f"Back to datetime: {datetime.fromtimestamp(ts)}")
```

```python
from datetime import date

# Practical example — Age calculator
def calculate_age(birthdate: date) -> dict:
    """Return age in years, and days until next birthday."""
    today = date.today()
    age = today.year - birthdate.year
    if (today.month, today.day) < (birthdate.month, birthdate.day):
        age -= 1

    # Days until next birthday
    next_bday = date(today.year, birthdate.month, birthdate.day)
    if next_bday < today:
        next_bday = date(today.year + 1, birthdate.month, birthdate.day)
    days_left = (next_bday - today).days

    return {"age": age, "days_until_birthday": days_left, "next_birthday": next_bday}


result = calculate_age(date(2007, 8, 27))
print(f"Age              : {result['age']} years")
print(f"Next birthday    : {result['next_birthday']}")
print(f"Days to go       : {result['days_until_birthday']} days")
```
