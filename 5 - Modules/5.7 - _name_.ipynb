## 7. `__name__ == "__main__"`

Every Python file has a built-in `__name__` attribute:

| Scenario | Value of `__name__` |
|----------|---------------------|
| File is **run directly** (`python myfile.py`) | `"__main__"` |
| File is **imported** as a module | The module's filename (e.g. `"myfile"`) |

### Why this pattern matters
The guard `if __name__ == "__main__":` lets you:
- Run **test/demo code** when executing a file directly
- **Suppress** that code when the file is imported as a library
- Write files that work both as **scripts** and **importable modules**

```
mymodule.py
├── def helper_func()  → always available on import
└── if __name__ == "__main__":  → only runs when executed directly
        main()
```

````python
# Inside a Jupyter notebook, __name__ is always '__main__'
print("__name__ in this notebook:", __name__)  # __main__

if __name__ == "__main__":
    print("This is the entry point — running directly!")
else:
    print("This file was imported as a module")
````

````python
# Write a module that behaves differently when imported vs run directly
calculator_src = '''
"""calculator.py — A reusable calculator module with a demo entry point."""


def add(a, b):      return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b


def _run_demo():
    """Demo function — only called when run directly."""
    print("=== Calculator Demo ===")
    pairs = [(10, 3), (7, 2), (100, 4)]
    for a, b in pairs:
        print(f"  {a} + {b} = {add(a, b)}")
        print(f"  {a} - {b} = {subtract(a, b)}")
        print(f"  {a} * {b} = {multiply(a, b)}")
        print(f"  {a} / {b} = {divide(a, b):.2f}")
        print()


if __name__ == "__main__":
    # This block runs ONLY when you do: python calculator.py
    # It is SKIPPED when you do: import calculator
    _run_demo()
'''

with open('calculator.py', 'w') as f:
    f.write(calculator_src)

print("calculator.py written ✓")
````

````python
# Scenario A — Import the module (demo does NOT run)
import calculator

print("calculator.__name__:", calculator.__name__)  # 'calculator'  (NOT '__main__')
print()

# Only the functions are available — no demo output
print("10 + 5 =", calculator.add(10, 5))
print("10 / 4 =", calculator.divide(10, 4))
````

````python
# Scenario B — Run the file directly (demo DOES run)
!python calculator.py
````

````python
# Best practice pattern used in real projects:
real_script_src = '''
"""A well-structured script following the __main__ pattern."""

import sys


def process(data: list) -> list:
    """Core logic — importable and testable."""
    return [x * 2 for x in data if x > 0]


def main():
    """Entry point — orchestrates the script."""
    raw = [1, -2, 3, -4, 5]
    result = process(raw)
    print("Input :", raw)
    print("Output:", result)
    return 0  # exit code 0 = success


if __name__ == "__main__":
    sys.exit(main())
'''

with open('real_script.py', 'w') as f:
    f.write(real_script_src)

!python real_script.py

# Cleanup
import os
for f in ['calculator.py', 'real_script.py']:
    if os.path.exists(f): os.remove(f)
print("\nCleanup done ✓")
````
