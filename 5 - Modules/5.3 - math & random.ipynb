## 3. `math` & `random`

### `math` - Mathematical Functions
Provides **deterministic** (same output every time) math functions and constants. Works with real numbers; for complex numbers use `cmath`.

| Category | Functions |
|----------|-----------|
| Constants | `pi`, `e`, `tau`, `inf`, `nan` |
| Rounding | `ceil()`, `floor()`, `trunc()`, `fabs()` |
| Power/Log | `sqrt()`, `pow()`, `exp()`, `log()`, `log2()`, `log10()` |
| Trig | `sin()`, `cos()`, `tan()`, `radians()`, `degrees()` |
| Combinatorics | `factorial()`, `gcd()`, `lcm()`, `comb()`, `perm()` |

### `random` - Pseudo-Random Numbers
| Function | What it returns |
|----------|-----------------|
| `random()` | float in `[0.0, 1.0)` |
| `uniform(a, b)` | float in `[a, b]` |
| `randint(a, b)` | int in `[a, b]` (inclusive) |
| `choice(seq)` | single random element |
| `sample(seq, k)` | k unique elements (no replacement) |
| `choices(seq, k=n)` | k elements (with replacement) |
| `shuffle(seq)` | shuffle list in-place |
| `seed(n)` | fix seed for reproducibility |

> ⚠️ `random` is **not** cryptographically secure. Use the `secrets` module for passwords and tokens.

## math

```python
import math
```

```python
# Constants
print(f"pi    = {math.pi}")
print(f"e     = {math.e}")
print(f"tau   = {math.tau}")
print(f"inf   = {math.inf}")
```

```python
# Rounding
print(f"\nceil(4.1)  = {math.ceil(4.1)}")   # 5
print(f"floor(4.9) = {math.floor(4.9)}")   # 4
print(f"fabs(-3.5) = {math.fabs(-3.5)}")   # 3.5
```

```python
# Power & roots
print(f"\nsqrt(144)     = {math.sqrt(144)}")      # 12.0
print(f"pow(2, 10)    = {math.pow(2, 10)}")      # 1024.0
print(f"log(100, 10)  = {math.log(100, 10)}")    # 2.0
print(f"log2(8)       = {math.log2(8)}")         # 3.0
```

```python
# Trigonometry - always uses RADIANS
angle_deg = 45
angle_rad = math.radians(angle_deg)
```

```python
print(f"sin(45°) = {math.sin(angle_rad):.4f}")   # 0.7071
print(f"cos(45°) = {math.cos(angle_rad):.4f}")   # 0.7071
print(f"degrees(π) = {math.degrees(math.pi)}")   # 180.0
```

```python
# Combinatorics
print(f"\nfactorial(5) = {math.factorial(5)}")    # 120
print(f"gcd(48, 18)  = {math.gcd(48, 18)}")      # 6
print(f"lcm(4, 6)    = {math.lcm(4, 6)}")        # 12
print(f"C(10,3)      = {math.comb(10, 3)}")       # 120  — combinations
print(f"P(10,3)      = {math.perm(10, 3)}")       # 720  — permutations
```

## random

```python
import random
```

```python
# Fix seed for reproducibility (great for testing)
random.seed(42)
```

```python
# Floats
print("random()        =", random.random())         # float [0, 1)
print("uniform(1, 10)  =", random.uniform(1, 10))   # float [1, 10]
```

```python
# Integers
print("randint(1, 6)   =", random.randint(1, 6))    # inclusive — like a die
print("randrange(0,10,2)=", random.randrange(0, 10, 2))  # 0,2,4,6,8
```

```python
# Sequences
cards = ['Ace', 'King', 'Queen', 'Jack', '10', '9']
print("\nchoice()  :", random.choice(cards))           # 1 random card
print("sample(k=3):", random.sample(cards, 3))         # 3 unique cards
print("choices(k=3):", random.choices(cards, k=3))     # 3 (may repeat)
```

```python
# Shuffle in-place
deck = list(range(1, 14))
random.shuffle(deck)
print("\nShuffled deck:", deck)
```

```python
# Weighted choices — 70% chance of 'Hit', 30% 'Miss'
results = random.choices(['Hit', 'Miss'], weights=[70, 30], k=10)
print("Weighted results:", results)
print(f"Hits: {results.count('Hit')}/10")
```

```python
# Simulating a dice roll 1000 times
from collections import Counter
rolls = [random.randint(1, 6) for _ in range(1000)]
freq = Counter(rolls)
print("\nDice frequency (1000 rolls):")
for face in sorted(freq):
    bar = '#' * (freq[face] // 5)
    print(f"  {face}: {bar} ({freq[face]})")
```
