## 6. `pip` & Virtual Environments

`pip` is Python's package installer. It downloads libraries from **PyPI** (Python Package Index - pypi.org).

### Most-used `pip` commands (run in terminal)
```bash
pip install requests            # install latest
pip install requests==2.28.0    # install exact version
pip install requests>=2.25.0    # minimum version
pip install --upgrade requests  # upgrade
pip uninstall requests          # remove
pip list                        # all installed packages
pip show requests               # info about one package
pip freeze > requirements.txt   # export all versions
pip install -r requirements.txt # install from file
```

### Virtual Environments — Why?
Without venvs, all projects share the same global packages → **version conflicts**.

```bash
# Create
python -m venv venv

# Activate
source venv/bin/activate        # macOS / Linux
venv\Scripts\activate           # Windows

# Install inside venv (isolated!)
pip install requests

# Deactivate when done
deactivate
```

> 💡 Always create a venv **per project** and add `venv/` to your `.gitignore`.

````python
# Demonstrate creating a venv via Python's subprocess module
import subprocess, sys, os

print("--- Creating a virtual environment ---")
result = subprocess.run(
    [sys.executable, '-m', 'venv', 'demo_venv'],
    capture_output=True, text=True
)
if result.returncode == 0:
    print("demo_venv created ✓")
    print("Contents:", os.listdir('demo_venv'))
else:
    print("Error:", result.stderr)

# Cleanup
import shutil
shutil.rmtree('demo_venv', ignore_errors=True)
print("Cleaned up demo_venv ✓")
````
