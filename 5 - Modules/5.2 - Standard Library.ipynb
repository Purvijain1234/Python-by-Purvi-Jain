## 2. Standard Library - `os` & `sys`

Python ships with a **rich standard library** - hundreds of built-in modules requiring no installation.

### `os` - Operating System Interface
| Category | Key functions |
|----------|---------------|
| Directory | `getcwd()`, `chdir()`, `listdir()`, `makedirs()` |
| Path | `path.join()`, `path.exists()`, `path.basename()`, `path.dirname()` |
| Environment | `environ.get()`, `environ[]` |
| File ops | `remove()`, `rename()`, `stat()` |

### `sys` - Python Interpreter Interface
| Attribute / Function | Purpose |
|----------------------|---------|
| `sys.version` | Python version string |
| `sys.argv` | Command-line arguments list |
| `sys.path` | Module search directories |
| `sys.exit()` | Exit the interpreter |
| `sys.stdout / stderr` | Standard I/O streams |

## os

```python
import os

# Current working directory
print("CWD:", os.getcwd())
```

```python
# List files (first 8 entries)
print("Files:", os.listdir('.')[:8])
```

```python
# Build cross-platform paths safely
path = os.path.join('home', 'user', 'documents', 'report.txt')
print("Path:", path)  # home/user/documents/report.txt  (or \ on Windows)
```

```python
# Path utilities
sample = '/home/user/documents/report.txt'
print(os.path.basename(sample))    # report.txt
print(os.path.dirname(sample))     # /home/user/documents
print(os.path.splitext(sample))    # ('/home/user/documents/report', '.txt')
print(os.path.exists(sample))      # False
```

```python
import os

# Create nested directories
os.makedirs('test_project/data/raw', exist_ok=True)  # exist_ok avoids error if exists
print("Created:", os.listdir('test_project'))

# Environment variables
home = os.environ.get('HOME', 'Not set')
print("HOME:", home)

# Set a temporary env variable (only for this process)
os.environ['APP_ENV'] = 'development'
print("APP_ENV:", os.environ.get('APP_ENV'))

# Cleanup
import shutil
shutil.rmtree('test_project')
print("cleanup done")
```

## Sys

```python
import sys

# Python version and platform
print("Version:", sys.version)
print("Version info:", sys.version_info)
print("Platform:", sys.platform)  # 'linux', 'darwin', 'win32'
```

```python
# Command-line args (in a real script: python script.py arg1 arg2)
print("argv:", sys.argv)          # [script_name, arg1, arg2, ...]
```

```python
# Module search path (Python looks here for imports, in order)
print("\nFirst 3 paths in sys.path:")
for p in sys.path[:3]:
    print(" ", p)
```

```python
# Standard streams
sys.stdout.write("Writing directly to stdout\n")
sys.stderr.write("This is a warning message\n")
```
