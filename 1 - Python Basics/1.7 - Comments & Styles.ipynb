## 7. Comments & Style

Good code is **readable** code. Python has two types of comments:

### Single-line comments
Start with `#`. Everything after `#` on that line is ignored by Python.

### Docstrings
Triple-quoted strings (`"""..."""`). Used to document **modules, functions, classes, and methods**. They are stored as the `__doc__` attribute and used by tools like `help()`.

### PEP 8 — Python's style guide (key rules)
| Rule | Good | Bad |
|------|------|-----|
| Indentation | 4 spaces | tabs or 2 spaces |
| Variable names | `snake_case` | `CamelCase`, `mixedCase` |
| Constants | `UPPER_CASE` | `myConstant` |
| Max line length | 79 characters | longer lines |
| Spaces around operators | `x = 1 + 2` | `x=1+2` |
| Blank lines | 2 between top-level defs | no blank lines |

```python
# PEP 8 compliant variable names
user_name = "Purvi"       # snake_case  ✓
MAX_RETRIES = 3            # constant in UPPER_CASE  ✓
is_logged_in = False
```

```python
def greet_user(name):
    """
    Return a greeting string for the given user name.

    Args:
        name (str): The user's name.

    Returns:
        str: A personalised greeting.
    """
    return f"Hello, {name}! Welcome to Python."
    
```

```python
# Access the docstring
print(greet_user.__doc__)
```

```python
# Calling the function
message = greet_user(user_name)
print(message)
```
