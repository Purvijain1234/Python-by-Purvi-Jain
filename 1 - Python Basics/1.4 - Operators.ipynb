## 4. Operators

Python has five groups of operators:

| Group | Operators |
|-------|-----------|
| Arithmetic | `+` `-` `*` `/` `//` `%` `**` |
| Comparison | `==` `!=` `>` `<` `>=` `<=` |
| Logical | `and` `or` `not` |
| Assignment | `=` `+=` `-=` `*=` `/=` `//=` `%=` `**=` |
| Identity / Membership | `is` `is not` `in` `not in` |

### Key distinctions
- `/` always returns a **float** (`7/2` → `3.5`)
- `//` is **floor division** - rounds down to nearest int (`7//2` → `3`)
- `%` is the **modulo** (remainder) - very useful for even/odd checks
- `**` is **exponentiation** (`2**10` → `1024`)

```python
# Arithmetic
print(10 + 3)   # 13
print(10 - 3)   # 7
print(10 * 3)   # 30
print(10 / 3)   # 3.3333...
print(10 // 3)  # 3  (floor division)
print(10 % 3)   # 1  (remainder)
print(2 ** 8)   # 256
```

```python
# Comparison  (returns True or False)
print(5 == 5)   # True
print(5 != 3)   # True
print(10 > 20)  # False
```

```python
# Logical
print(True and False)   # False
print(True or False)    # True
print(not True)         # False
```

```python
# Assignment shorthand
counter = 0
counter += 5    # same as counter = counter + 5
print(counter)  # 5
```

```python
# Membership
fruits = ["apple", "banana", "mango"]
print("apple" in fruits)      # True
print("grape" not in fruits)  # True
```

```python
# Identity  (checks same object in memory, not just equal value)
a = None
print(a is None)   # True  (correct way to check for None)
print(a == None)   # True  (works but 'is' is preferred)
```
