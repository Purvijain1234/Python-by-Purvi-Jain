## 3. Input & Output

### Output - `print()`
The `print()` function displays values to the screen. It accepts multiple arguments and has two useful keyword arguments:
- `sep` - separator between values (default is a space `" "`)
- `end` - what to print at the end (default is newline `"\n"`)

### Input - `input()`
`input()` reads a line of text typed by the user and **always returns a string**. If you need a number, you must cast it explicitly.

> In notebooks, `input()` opens an interactive prompt box. Run the cell and type your value.

```python
# Basic print
print("Hello, World!")
```

```python
# Multiple values with custom separator
print("Python", "is", "fun", sep="-")   # Python-is-fun
```

```python
# Suppress the newline
print("Loading", end="...")
print("Done")
```

```python
# Print different types together
score = 95
print("Your score is:", score, "out of 100")
```

```python
# Input
name = input("Enter your name: ")
print("Hello,", name)
```

```python
age = int(input("Enter your age: "))   # cast to int!
print("Next year you will be", age + 1)
```
