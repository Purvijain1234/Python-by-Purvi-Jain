## 1. Variables & Data Types

A **variable** is a named container that stores a value. Python is **dynamically typed** - you don't need to declare the type; Python figures it out automatically.

### Core built-in types
| Type | Keyword | Example |
|------|---------|--------|
| Integer | `int` | `42` |
| Float | `float` | `3.14` |
| String | `str` | `"hello"` |
| Boolean | `bool` | `True` / `False` |
| None | `NoneType` | `None` |

> Use `type()` to check the type of any variable at runtime.

```python
# Integer
age = 21
print(type(age), age)
```

```python
# Float
price = 99.99
print(type(price), price)
```

```python
# String
name = "Purvi"
print(type(name), name)
```

```python
# Boolean
is_student = True
print(type(is_student), is_student)
```

```python
# NoneType
result = None
print(type(result), result)
```

```python
# Multiple assignment in one line
x, y, z = 1, 2.5, "three"
print(x, y, z)
```
