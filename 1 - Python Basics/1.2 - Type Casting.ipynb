## 2. Type Casting

**Type casting** (also called type conversion) means converting a value from one type to another.

- `int()` — converts to integer (truncates decimals, doesn't round)
- `float()` — converts to float
- `str()` — converts to string
- `bool()` — converts to boolean

### Truthy vs Falsy
When converting to `bool`, the following are **falsy** (become `False`):
`0`, `0.0`, `""`, `None`, `[]`, `{}`, `()`  
Everything else is **truthy** (`True`).

```python
# str -> int
num_str = "42"
num_int = int(num_str)
print(type(num_int), num_int)
```

```python
# float -> int  (truncates, does NOT round)
pi = 3.99
print(int(pi))       # 3, not 4
```

```python
# int -> float
print(float(7))      # 7.0
```

```python
# int / float -> str
print(str(100))      # '100'
print(str(3.14))     # '3.14'
```

```python
# Truthy / Falsy
print(bool(0))       # False
print(bool(""))      # False
print(bool(None))    # False
print(bool(42))      # True
print(bool("hello")) # True
```
