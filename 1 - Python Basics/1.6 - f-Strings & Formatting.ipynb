## 6. f-strings & Formatting

**f-strings** (formatted string literals, introduced in Python 3.6) are the modern, preferred way to embed expressions inside strings. Prefix the string with `f` and put any expression inside `{}`.

### Format specifiers inside `{}`
```
{value:.2f}   → 2 decimal places
{value:,}     → comma as thousands separator
{value:>10}   → right-align in field of width 10
{value:<10}   → left-align
{value:^10}   → center-align
{value:05d}   → zero-padded integer
```

> Python 3.8+ supports **self-documenting expressions** with `=` inside `{}`:  
> `f"{score=}"` prints `score=95` - very handy for quick debugging.

````python
name = "Purvi"
age = 21
score = 95.678
````

````python
# Basic f-string
print(f"Name: {name}, Age: {age}")
````

````python
# Expressions inside {}
print(f"Next year I will be {age + 1}")
print(f"Is adult: {age >= 18}")
````

````python
# Float formatting
print(f"Score: {score:.2f}")       # 95.68
print(f"Score: {score:.0f}")       # 96
````

````python
# Thousands separator
salary = 1200000
print(f"Salary: {salary:,}")       # 1,200,000
````

````python
# Alignment & padding
for item in [("Apple", 50), ("Banana", 120), ("Mango", 80)]:
    print(f"{item[0]:<10} {item[1]:>5}")
````

````python
# Zero padding
for i in range(1, 6):
    print(f"File_{i:03d}.txt")     # File_001.txt, File_002.txt ...
````

````python
# Debug with = 
x = 42
print(f"{x=}")                     # x=42
````
