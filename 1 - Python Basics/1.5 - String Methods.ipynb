## 5. String Methods

Strings in Python are **immutable sequences of characters**. They come with a rich set of built-in methods — none of them modify the original string; they always return a new one.

### Most-used methods
| Method | What it does |
|--------|--------------|
| `.upper()` / `.lower()` | Change case |
| `.strip()` | Remove leading/trailing whitespace |
| `.split(sep)` | Split into a list by separator |
| `.join(iterable)` | Join a list into a string |
| `.replace(old, new)` | Replace substring |
| `.find(sub)` | Return index of first match (-1 if not found) |
| `.startswith()` / `.endswith()` | Check prefix/suffix |
| `.count(sub)` | Count occurrences |
| `len(s)` | Length of string (built-in, not a method) |

```python
s = "  Hello, Python World!  "
```

```python
print(s.strip())           # remove whitespace
print(s.strip().lower())   # chaining methods
print(s.strip().upper())
```

```python
sentence = "the quick brown fox"
```

```python
print(sentence.title())    # The Quick Brown Fox
print(sentence.capitalize()) # The quick brown fox
```

```python
# split & join
words = sentence.split(" ")   # ['the', 'quick', 'brown', 'fox']

print(words)
print("-".join(words))         # the-quick-brown-fox
```

```python
# replace
print(sentence.replace("fox", "dog"))
```
