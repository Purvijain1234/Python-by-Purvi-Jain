# 🔢 Project 1 - Calculator

## What it demonstrates
| Concept | Where used |
|---------|------------|
| Functions | Each operation is a separate function |
| Exception handling | Division by zero, invalid input |
| Dictionaries | Operator dispatch table |
| while loop | Continuous calculator session |
| f-strings | Formatted output |
| Type casting | `float(input(...))` |

## Features
- Basic arithmetic: `+`, `-`, `*`, `/`, `//`, `%`, `**`
- Memory store / recall (M+, MR, MC)
- History of last 5 calculations
- Graceful error handling for all invalid inputs
- Expression evaluator for chained calculations

## How it works
```
User enters: 10 + 5
         ↓
  parse_expression()  →  ('10', '+', '5')
         ↓
  dispatch table  →  OPERATIONS['+']  →  add(10, 5)
         ↓
  result = 15.0   →  stored in history
```

````python
# ============================================================
#  PROJECT 1 — CALCULATOR  (full implementation)
# ============================================================

from collections import deque

# ---- Core operations ----
def add(a, b):       return a + b
def subtract(a, b):  return a - b
def multiply(a, b):  return a * b
def power(a, b):     return a ** b
def floor_div(a, b): return a // b
def modulo(a, b):    return a % b

def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return a / b

# Dispatch table — maps operator symbol → function
OPERATIONS = {
    "+":  (add,       "Addition"),
    "-":  (subtract,  "Subtraction"),
    "*":  (multiply,  "Multiplication"),
    "/":  (divide,    "Division"),
    "//": (floor_div, "Floor Division"),
    "%":  (modulo,    "Modulo"),
    "**": (power,     "Exponentiation"),
}

# ---- Calculator state ----
class Calculator:
    """Stateful calculator with memory and history."""

    def __init__(self):
        self.memory  = 0.0
        self.history = deque(maxlen=5)   # keep last 5 results
        self.last    = None

    def calculate(self, a: float, op: str, b: float) -> float:
        """Perform operation and record in history."""
        if op not in OPERATIONS:
            raise ValueError(f"Unknown operator: {op!r}")
        func, name = OPERATIONS[op]
        result = func(a, b)
        entry = f"{a} {op} {b} = {result}"
        self.history.append(entry)
        self.last = result
        return result

    def memory_add(self, value):
        self.memory += value
        print(f"  M+ : memory = {self.memory}")

    def memory_recall(self):
        print(f"  MR : memory = {self.memory}")
        return self.memory

    def memory_clear(self):
        self.memory = 0.0
        print("  MC : memory cleared")

    def show_history(self):
        print("  📋 History (last 5):")
        if not self.history:
            print("    (empty)")
        for i, entry in enumerate(self.history, 1):
            print(f"    {i}. {entry}")

    def parse_and_compute(self, expression: str) -> float | None:
        """Parse 'a op b' string and compute."""
        # Try two-char operators first (// and **)
        for op in ["//", "**", "+", "-", "*", "/", "%"]:
            parts = expression.split(op)
            if len(parts) == 2:
                try:
                    a = float(parts[0].strip())
                    b = float(parts[1].strip())
                    return self.calculate(a, op, b)
                except (ValueError, ZeroDivisionError) as e:
                    print(f"  ⚠️  {e}")
                    return None
        print("  ⚠️  Could not parse expression.")
        return None

print("✅ Calculator class defined successfully.")
````

````python
# ---- Demo — simulate a full calculator session ----

calc = Calculator()

demo_expressions = [
    "10 + 5",
    "100 / 4",
    "2 ** 10",
    "17 % 5",
    "99 // 10",
    "25 * 4",
    "10 / 0",       # error case
    "abc + 3",      # invalid input
]

print("=" * 45)
print("          🔢  PYTHON CALCULATOR")
print("=" * 45)

for expr in demo_expressions:
    print(f"\n  Input : {expr}")
    result = calc.parse_and_compute(expr)
    if result is not None:
        # Format: remove trailing .0 for whole numbers
        formatted = int(result) if result == int(result) else round(result, 6)
        print(f"  Result: {formatted}")

# Memory operations
print("\n" + "-" * 45)
print("  Memory operations:")
calc.memory_add(calc.last)    # store last result
calc.memory_add(500)
calc.memory_recall()
calc.memory_clear()

# History
print()
calc.show_history()

# Chain calculation using last result
print("\n  Chained calculation:")
r1 = calc.calculate(50, "+", 50)
r2 = calc.calculate(r1, "*", 3)
r3 = calc.calculate(r2, "/", 4)
print(f"  (50 + 50) * 3 / 4 = {r3}")

print("\n" + "=" * 45)
````
